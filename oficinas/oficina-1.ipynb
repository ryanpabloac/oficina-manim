# <h2 id='primeira-animacao'> 1.4 🎬 A "Anatomia" de uma Cena em Manim

---
A estrutura de um código Manim não é apenas funcional: ela segue uma coreografia lógica que separa a **existência** da **ação**.

1.   **Os Bastidores (Definição dos Objetos em cena)**:

Nesta fase, você atua como o **cenógrafo**. Você define as propriedades matemáticas e estéticas, mas nada é visível ainda. É o momento da precisão.

Aqui criamos uma `class` que herda todos os comportamentos de `Scene` (ou de outra cena especificada, como por exemplo `Slides`) e utilizamos o método `def construct()` para definir os objetos em cena.


2. **O Palco (A Chamada dos Objetos para a Cena)**:

Aqui você assume o papel de **diretor**. Através de métodos `self.play()` ou `self.add()` você traz os objetos da "escuridão" para a luz, definindo como eles devem aparecer.   


Para testar, execute o código abaixo e veja como a mágica acontece.




```python
# @title
%%manim -qm -v WARNING Teste

class Teste(Scene): # obrigatório
    def construct(self): # obrigatório

        # --- CONFIGURAÇÕES (Opcional) ---
        # self.camera.background_color = "#5F9EA0"

        # --- BASTIDORES (Definição dos Objetos) ---
        # Aqui o objeto é configurado, escalonado e colorido na memória
        textoIME = Text('Instituto de Matemática e Estatística - IME').scale(0.8).set_color(WHITE)

        # --- PALCO (Ação e Apresentação) ---
        # O momento em que o objeto ganha vida na tela
        self.play(Write(textoIME))

        # Tempo de espera para contemplação da cena
        self.wait(6)
```

## Decifrando o Código

---

* `%%manim`:  um "*magic command*" usado em notebooks Jupyter (como no Google Colab) para rodar animações do Manim diretamente dentro da célula do notebook.

- `-v WARNING`: significa que o Manim só mostrará avisos e mensagens de erro enquanto estiver executando, ignorando outros tipos de saída de log, como informações gerais e mensagens de depuração.

* `self.camera`: refere-se à câmera da cena (a "visão" que o Manim usa para renderizar).

* `background_color`: é o atributo da câmera que define a cor do plano de fundo (o fundo da animação deixará de ser preto, que é o padrão).

* `self.add( )`: o objeto surge imediatamente, sem efeito.

* `self.play( )`: o objeto será animado.

* `self.wait(n)`: suspende a execução de novas animações por $n$ segundos, mantendo todos os objetos atuais na tela.

Os parâmetros utilizados para alterar configurações de renderização são:

- `-p` (preview) - abre o vídeo em um programa instalado no computador.

- `-q` (quality) - especifica qualidade da animação:
    
    - `-l`(low) - Qualidade baixa.

    - `-m`(medium) - Qualidade média.

    - `-h`(high) -  Qualidade alta.

    - `-k`(4k) - Qualidade de 4k.

- `-s` - renderiza o último frame (quadro) em uma imagem.

- `-i` -  renderiza um gif.



#1.5  Alguns Elementos da Cena - *Mobject*

---

No Manim, tudo o que aparece na cena (como textos, fórmulas matemáticas, formas geométricas, gráficos ou imagens) é chamado de *objeto matemático*, ou simplesmente *Mobject* (do inglês mathematical object).

Esses elementos são os "blocos de construção" da animação. Sempre que você quiser mostrar algo visual na tela, será necessário criar e manipular um Mobject.

Vejamos alguns Mobjects frequentemente utilizados.


##1.5.1 Comando `Text`

---

O comando `Text` é usado para criar textos simples (não matemáticos) nas animações.



```python
# @title
%%manim -qm -v WARNING Cena1

class Cena1(Scene):
    def construct(self):

        # --- BASTIDORES (Definição dos Objetos) ---
        # Aqui o objeto T1 é criado na memória, mas ainda não é visível
        T1 = Text('Instituto de Matemática e Estatística')

        # --- PALCO (Ação e Apresentação) ---
        # Chamamos o objeto para a cena. O método add exibe o objeto instantaneamente
        self.add(T1)

        # Tempo de espera para visualização
        self.wait(2)
```


Podemos mudar a cor do texto usando o argumento `color` ou criar gradientes de cores usando `gradient`.  Também é possível mudar a formatação do texto para *itálico* usando o argumento `slant` com o parâmetro `ITALIC`. Para mudar a fonte do texto usamos o argumento `font`.

No script abaixo, definimos mais de uma cena para exemplificar os argumentos mencionados acima. Para renderizar uma cena específica, basta informar o nome da classe na linha de comando.

```python
# @title
%%manim -qk -v WARNING Texto3

# --- CENA 1: Introdução e Imagem ---
class Texto1(Scene):
    def construct(self):
        # --- BASTIDORES ---
        # self.camera.background_color = BLUE
        # image = ImageMobject("caminho/para/sua/imagem.png")

        T1 = Text('Instituto de Matemática e Estatística', color=BLUE_A, font='cursive')

        # --- PALCO ---
        # self.add(image)
        self.add(T1)


# --- CENA 2: Cores e Gradientes ---
class Texto2(Scene):
    def construct(self):
        # --- BASTIDORES ---
        T2 = Text('Instituto de Matemática e Estatística', gradient=(YELLOW_C, RED, BLUE))
        # Também é possível mudar a cor padrão usando o set_color_by_gradiente:
        # T2.set_color_by_gradient(BLUE_C, GREEN_D, PINK)

        # --- PALCO ---
        self.add(T2)


# --- CENA 3: Estilos (Itálico) ---
class Texto3(Scene):
    def construct(self):
        # --- BASTIDORES ---
        T3 = Text('Instituto de Matemática e Estatística', slant=ITALIC)
        # É possível mudar a cor padrão usando o argumento color:
        # texto = Text('Texto vermelho', color=RED)

        # --- PALCO ---
        self.add(T3)


# --- CENA 4: Fontes Específicas ---
class Texto4(Scene):
    def construct(self):
        # --- BASTIDORES ---
        T4 = Text('Instituto de Matemática e Estatística', font='DejaVu Sans', color=RED_B)
        # Também é possível mudar a cor padrão usando o argumento color.
        # texto = Text('Texto vermelho', color=RED)

        # --- PALCO ---
        self.add(T4)
```

##1.5.2 Comando `MarkupText`

---

Quando quisermos aplicar diferentes estilos em partes do mesmo texto usamos o `MarkupText`. Com isso, evitamos criar vários `Text()`. Os principais marcadores que podemos usar com `MarkupText` são:

* `<b>Texto</b>`: Negrito

* `<i>Texto</i>`: Itálico

* `<u>Texto</u>`: Sublinhado

* `<s>Texto</s>`: Texto riscado
* `<span fgcolor="cor">texto</span>`:  Muda a cor da letra
* `<span bgcolor="cor">texto</span>`:  Muda a cor do fundo da palavra, criando um efeito de caneta marca-texto.

```python
# @title
%%manim -qm -v WARNING CitacaoMarkup

class CitacaoMarkup(Scene):
    def construct(self):

        # --- BASTIDORES (Definição dos Objetos) ---
        # Definimos o texto com as marcações de estilo e cores
        texto = MarkupText('A <i><span bgcolor="blue">Matemática</span></i> é o <b><span fgcolor="red">coração </span></b> de todas as <u><span fgcolor="yellow">ciências </span></u>.',font_size=38)

        # --- PALCO (Ação e Apresentação) ---
        # Execução da animação de escrita
        self.play(Write(texto))

        # Tempo de espera para leitura
        self.wait(2)
```

## 1.5.3 Comando `MathTex` (Texto em LaTeX)

---

A partir deste momento, assumiremos que o leitor já possui familiaridade com LaTeX para a escrita de expressões matemáticas. Caso contrário, recomenda-se a leitura de [Apostila sobre LaTeX](https://www.facom.ufu.br/~william/Disciplinas%202015-2/BSI%20-%20GSI003%20-%20Introducao%20Sistemas%20Informacao/Template/ApostilaLaTex.pdf).

Para exibir fórmulas matemáticas na tela, usamos o comando `MathTex`, uma ferramenta específica para expressões matemáticas, permitindo renderizar símbolos e fórmulas (como frações, somatórios e integrais) com a formatação correta e bonita.




```python
# @title
%%manim -qm -v WARNING Latex1

class Latex1(Scene):
    def construct(self):

        # --- BASTIDORES (Definição dos Objetos) ---
        # self.camera.background_color = TEAL_D

        # Definição das fórmulas principais com gradiente
        T1 = MathTex(
            r"\int_{a}^{b} \frac{x+6}{x^2+12x}\; dx&= \left[\frac{1}{2}\ln|x^2+12x|\right]_a^b\\ "
            r"&\sum_{a_n}^{\infty}a_n \\ "
            r"&f'(x)= \lim_{t\to x}\frac{f(t)-f(x)}{t-x}"
        )
        T1.set_color_by_gradient(RED, GREEN, PURE_GREEN, PINK)

        # Definição e posicionamento do segundo texto
        T2 = MathTex(r"\lim_{x\to a}f(x)=f(a)", color=ORANGE)
        T2.move_to(UP * 3)
        # T2.next_to(T1, UP)

        # --- PALCO (Ação e Apresentação) ---
        # self.add(T1)
        self.play(Create(T1), run_time=3)
        self.wait(1)

        # Adição do segundo texto ao palco
        self.add(T2)
        self.wait(1)
```

Agora, se quisermos renderizar qualquer tipo de texto, matemático ou não, usamos o comando`Tex`. Aqui, como em ambiente LaTex, os símbolos matemáticos devem ser colocados entre `$`.



```python
# @title
%%manim -qm -v WARNING Latex2

class Latex2(Scene):
    def construct(self):

        # --- BASTIDORES (Definição dos Objetos) ---
        # T1 = Tex(r"Para todo $x$ pertencente ao conjunto dos números reais: $\forall x \in \mathbb{R}$.")
        T1 = Tex(r"Para todo $x$ pertencente ao conjunto dos números reais: $\forall x \in \mathbb{R}$.").scale(0.8)
        # T1 = Tex(r"Para todo $x$ pertencente ao conjunto dos números reais: $\forall x \in \mathbb{R}$.", font_size=40)

        # --- PALCO (Ação e Apresentação) --
        self.play(FadeIn(T1))

        # Tempo de espera para leitura do símbolo lógico
        self.wait(5)
```

```python
# @title
%%manim -qm -v WARNING FormulaCauchy

class FormulaCauchy(Scene):
    def construct(self):
        # --- BASTIDORES (Definição dos Objetos) ---
        formula = MathTex(r"[z^n]f(z) = \frac{1}{2\pi i}\oint_{\gamma} \frac{f(z)}{z^{n+1}}~dz")
        #formula.set_color_by_gradient(BLUE,RED,GREEN,LIGHT_PINK)

        # Criando a caixa (SurroundingRectangle é um objeto independente)
        box = SurroundingRectangle(formula, color=WHITE, buff=0.2)  #No Manim, o parâmetro buff refere-se à distância de "buffer" ou margem de espaço entre dois objetos,
        #como, por exemplo, o espaço entre o texto e o contorno de uma caixa ou o espaço entre formas geométricas.


        # --- PALCO (Ações e Animações) ---
        # 1. Escreve a fórmula
        self.play(Write(formula), run_time=3)
        self.wait()

        # 2. Mostra a caixa
        self.play(Create(box))
        self.wait()

        # 3. Adiciona o fundo colorido à fórmula
        # Nota: add_background_rectangle é um método do objeto, não uma animação 'play' por padrão
        formula.add_background_rectangle(
            color=BLUE,
            opacity=0.5,
            buff=0.2
        )
        self.wait(3)
```

## 1.5.4 Texto com Cores Variadas


Podemos aplicar cores diferentes a partes específicas de um texto ou expressão matemática. Para isso, usamos o comando

`M1=MathTex(r"Texto0", r"Texto1",r"Texto2",...,r"Texto n")`.

Aqui, pensamos `M1` como um vetor de entradas, onde

`M1=(T0,T1,T2,...,Tn)` e então
`M1[0]=T0`; `M1[1]=T1`
; `M1[2]=T2`; `...`
 ; `M1[n+1]=Tn`.





```python
# @title

%%manim -qm M1
class M1(Scene):
    def construct(self):

        # --- CONFIGURAÇÕES (Opcional) ---
        self.camera.background_color = WHITE

        # --- BASTIDORES (Definição dos Objetos) ---
        # Definindo o conjunto de fórmulas M1 e suas propriedades
        M1 = MathTex(
            r"\lim_{x\to b}f(x)=f(b)\\",
            r"\int udv=uv-\int vdu \\",
            r"\int_a^b f(x)\; dx=F(b)-F(a)"
        )
        M1.scale(1)

        # Definindo a cor de cada entrada individualmente:
        M1[0].set_color(RED)
        M1[1].set_color(PURPLE)
        M1[2].set_color(GREEN)

        # Definindo o objeto M2 para a futura transformação
        M2 = MathTex(r"\ln(x^2+1)").to_edge(UP).set_color(BLACK)

        # --- PALCO (Ação e Apresentação) ---
        # Adiciona o conjunto inicial à cena
        self.add(M1)
        self.wait(3)

        # Executa a transformação da primeira fórmula para o novo texto
        self.play(Transform(M1[0], M2))
        self.wait(3)
```
