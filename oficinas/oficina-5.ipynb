# 5.2 Listas

Às vezes, queremos armazenar diversos elementos em uma única variável. Para isso, usamos **listas**. Para criá-las, usamos colchetes (`[ ]`) e inserimos os valores (objetos) separados por vírgulas, por exemplo:
1. Números naturais:  `[1, 2, 3, 4, 5, 6, 7, 8, 9]`
2. Cores: `[BLUE, PINK, RED]`
3. Funções: `[f_1,f_2,...,f_n]`
4. Elementos diferentes: `[BLUE, 3, f,...]`.

Para acessar um elemento de uma lista com $n$, usamos os índices $0$ à $n-1$. Por exemplo:

 * `Lista=[a_1,a_2,\cdot, a_n$]`
* `Lista[0]=a_1`
* `Lista[1]=a_2`
* `Lista[n-1]=a_n`.

Quando fazemos

```
Lista = [1, 2, 3]

print(*lista)
```

o `*` "desempacota" os elementos da lista. Isso significa que, em vez de imprimir a lista como `[1, 2, 3]`, o print recebe cada elemento separadamente como se fossem vários argumentos:
`1 2 3`.

# 5.3 Estruturas de controle.

As estruturas de controle permitem alterar o fluxo de execução de um programa, tomando decisões ou repetindo instruções conforme necessário. Utilizaremos o  `if`, `else`, `elif`, `while` e `for`.

## 5.3.1 Comandos `if`, `elif` e `else`

O comando **`elif`** (abreviação de *else if* ) é uma extensão da estrutura condicional. Com ele, é possível testar várias condições diferentes, sem precisar encaixar vários `if` e `else` dentro uns dos outros.

Assim, temos:

`if`: testa a primeira condição.

`elif`: testa uma nova condição, caso a anterior seja falsa.

`else`: executa quando nenhuma das condições anteriores é satisfeita.

Por exemplo:

```
n = 5

if n < 5:

print(f'{n} é menor que 5')

elif n == 5:

print(f'{n} é 5')

elif n > 5:

print(f'{n} é maior que 5')

else:

print(f'{n} não é um número')

```
  

````python
%%manim -qm if_elif_else

class if_elif_else(Scene):
    def construct(self):
        C=Circle(radius=0.3, color=BLUE,fill_opacity=1)
        R=Rectangle(width=2, height=3, color=RED)
        Q=Square(side_length=2, color=YELLOW)
        wt=0.5   # Define o tempo padrão de pausa
        num=4    # variável (guardando o valor na memória do computador para usar depois)

        #O operador % é o módulo (resto da divisão).
        if num % 3 ==0:
          self.add(C)
          self.wait(wt)
          self.play(C.animate.to_edge(UR))
          self.wait(wt)
        elif num % 2 ==0:
          self.add(R)
          self.wait(wt)
          self.play(R.animate.move_to([-3,1,0]))
          self.wait(wt)
        elif num % 5 ==0:
          self.add(Q)
          self.wait(wt)
          self.play(Q.animate.move_to([0,2,0]))
          self.wait(wt)
        else:
          self.add(Q)
          self.wait(wt)
          self.play(Q.animate.move_to([0,-2,0]))
          self.wait(wt)
````

## 5.3.2 Comando `for`

O comando `for` é uma estrutura de repetição usada para percorrer (iterar) cada elemento de uma sequência, como listas, strings ou outras coleções de dados.
Em outras palavras, ele permite executar um bloco de código para cada item presente na estrutura.

Na sua forma mais comum, o `for` é utilizado em conjunto com a palavra-chave `in`, que indica de onde os elementos serão retirados.

````python
%%manim -qm for_if_else

class for_if_else(Scene):
    def construct(self):
        C=Circle(radius=0.2, color=BLUE,fill_opacity=1)
        C2=Rectangle(width=0.2, height=0.3, color=RED,fill_opacity=1)

        wt=0.5 # Define o tempo padrão de pausa
        lista=[1,2,3,4,5,6,7,8]
        L=DashedLine([0,-2,0],[0,2,0])
        L2=DashedLine([-2,0,0],[2,0,0])
        #self.add(C2)
        self.add(L,L2)
        for num in lista:

            if num % 2 == 0:
              self.add(C)
              self.add(C2)
              self.wait(wt)
              self.play(C.animate.move_to([0,2,0]))
              self.play(C2.animate.move_to([2,0,0]))
              self.wait(wt)

   # Se não satisfeita, então redirecionada para o else

            else:
             self.add(C,C2)
             self.wait(wt)
             self.play(C.animate.move_to([0,-2,0]))
             self.play(C2.animate.move_to([-2,0,0]))
             self.wait(wt)


````

### 5.3.2.1 Criando listas com o laço `for`

Podemos criar uma lista de objetos (como círculos, textos, linhas, etc.) utilizando um laço `for`.  Isso é útil quando queremos gerar vários elementos de forma automática, em vez de criá-los manualmente um a um.

Além disso, podemos simplificar e otimizar o código utilizando uma *compreensão de lista* (list comprehension).

Imagine que criamos uma lista com vários círculos: $Cir = [c_1, c_2, \dots, c_n]$. Em vez de fazer um laço `for` tradicional que animaria um círculo de cada vez, podemos usar o operador asterisco `*` junto com a compreensão de lista para criar todos eles simultaneamente na tela:
```
self.play(*[Create(c) for c in Cir])
```



### 5.3.2.2  Exemplo 1 - Criando $n$ Objetos

# 5.4 Adicionando elementos a uma Lista (`.append`) e a um VGroup (`.add`)

#### 1. Adicionando elementos a uma lista com `.append`:

O método `.append()` serve para:

* Inserir elementos: Adiciona um novo objeto sempre no final da lista.

* Armazenar para o futuro: Guarda os objetos na memória para que possamos usá-los ou animá-los mais tarde.

Veja como funciona.
```
# 1. Cria uma lista vazia (ou com algum elemento) para começar a guardar os objetos
lista = []

# 2. Adiciona um círculo (ele entra na posição 0)
lista.append(Circle())

# 3. Adiciona um quadrado (ele vai para o final, após o círculo)
lista.append(Square())

# 4. Adiciona um triângulo (vai para o final, após o quadrado)
lista.append(Triangle())
```

#### 2. Adicionando elementos a um grupo com `.add`:

Enquanto o `.append()` pertence ao Python, o método `.add()` é uma ferramenta exclusiva do Manim. Ele serve para inserir objetos dentro de uma coleção gráfica chamada VGroup (Grupo de Vetores).

O `.add()` serve para:

* Agrupar geometricamente: Une vários objetos isolados em um único bloco. Se você mover, rotacionar ou redimensionar o VGroup, todos os objetos dentro dele obedecem juntos!

* Simplificar Animações: Em vez de dar self.play em 10 objetos separados, você coloca todos no grupo e anima o grupo inteiro com uma única linha de código.

Veja como funciona.

```

# 4. Adiciona o T2 ao mesmo grupo!
grupo.add(T2)
```

# 5.6 Updaters aplicados a objetos: pontos, figuras e linhas

___


No Manim, um `updater` é uma função que permite criar animações contínuas, dinâmicas e fluidas, indo muito além das transições estáticas ou lineares baseadas apenas no comando tradicional `self.play()`.

Nesta seção, exploraremos como aplicar os `updaters` para movimentar objetos de forma inteligente no plano cartesiano. A estratégia fundamental consiste em vincular as propriedades geométricas de uma figura (como a posição de um ponto, o raio de um círculo ou as coordenadas de uma linha) a um `ValueTracker`. Como esse rastreador dita o valor numérico em tempo real, qualquer alteração nele força o Manim a recalcular e reposicionar o objeto instantaneamente a cada quadro (frame), resultando em movimentos perfeitamente suaves e controlados matematicamente.

````python
# @title
%%manim -qm Updater_ponto
class Updater_ponto(Scene):
    def construct(self):
      ax=Axes(
      x_range=[-4,4,1],
      y_range=[-3,3,1],
      x_length=8,
      y_length=6,

      ).add_coordinates()
      self.add(ax)

      #Pontos
      OB=Dot().set_color(YELLOW_C)
      #Figuras
      #OB=Circle(radius=0.5).set_color(YELLOW_C)
      #OB=Triangle().set_color(YELLOW_C)
      #OB=Square(side_length=2).set_color(YELLOW_C)
      #OB=RegularPolygon(n=4).set_color(YELLOW_C)

      # Linha

      #OB=DashedLine(ax.c2p(0,0),ax.c2p(4,0)).set_color(YELLOW_C)


      Valor=ValueTracker(0)

      # Função updater:
      # Ela será chamada automaticamente a cada frame da animação
      def updater_OB(OB_movel):
        OB_movel.move_to(UP*Valor.get_value()) # move_to(...) reposiciona o objeto
                                               # Valor.get_value() pega o valor atual do ValueTracker

      # add_updater conecta a função ao objeto OB
      # A partir daqui, o Manim executa updater_OB(OB)
      # automaticamente em todos os frames

      OB.add_updater(updater_OB)
      # Adiciona o objeto na cena
      self.add(OB)

      self.play(Valor.animate.set_value(2), run_time=4, rate_func=smooth)
      self.wait()
       # rate_func controla a velocidade da animação
       #Alguns tipos de controles:
       # - smooth=Começa devagar, acelera, depois desacelera.
       # - linear = velocidade constante
       # - rush_into=Começa rápido e termina lento.
       # - rush_from= Começa lento e termina rápido.
       # - there_and_back=Vai e volta automaticamente.
       # - wiggle=Movimento tipo tremida.

      self.play(Valor.animate.set_value(0), run_time=4, rate_func=linear)
      self.wait()


````

##   5.6.2 A função `def updater(mob, dt):`

A função `def updater(mob, dt)` permite que o objeto ganhe "vida própria", atualizando suas propriedades quadro a quadro (frame por frame) com base no *tempo real decorrido*. É a ferramenta ideal para criar movimentos constantes,  como engrenagens girando, relógios correndo ou simulações físicas de partículas, que continuam rodando em segundo plano mesmo durante comandos simples de pausa, como o `self.wait()`.


💡 Diferença:

No `updater` anterior (sem dt): O objeto se move porque um valor numérico mudou.

No `updater` atual (com dt): O objeto se move porque o tempo do relógio passou.



````python
%%manim -qm DT

class DT(Scene):
    def construct(self):
        ax=NumberPlane()
        self.add(ax)

        bola = Circle(radius=0.2, color=YELLOW, fill_color=RED, fill_opacity=1)



        velocidade = 2
        #dt tempo entre frames

        def updater(bola, dt):

            # 2 unidades por segundo

            bola.shift(RIGHT * velocidade * dt)

        bola.add_updater(updater)

        self.add(bola)

        #Durante 3 segundos a bola anda aproximadamente:2×3=6 unidades

        self.wait(3)
````

## 5.6.2 Exemplo de `updaters` em segmentos

___
Neste exemplo, utilizamos `updaters` para modificar dinamicamente o comprimento de um segmento de reta. O valor controlado por um ValueTracker ajusta a posição do ponto final do segmento em cada frame da animação, permitindo que a reta cresça ou diminua de forma contínua.

Um detalhe importante é o uso do método `become()`. Ele substitui o objeto original por uma nova versão atualizada (neste caso, uma nova reta com extremidade modificada). Dessa forma, a linha não é apenas movida, mas reconstruída em cada instante de acordo com o valor atual do ValueTracker.

## 5.6.3 Exemplo de `updaters` em Pontos nos Gráficos

````python
%%manim -qm Updater_ponto_grafico

class Updater_ponto_grafico(Scene):
  def construct(self):


        ax = Axes(
            x_range=[0, 2*PI, PI/2],
            y_range=[-2, 2, 1],
        )
        labels = ax.get_axis_labels(x_label="x", y_label="f(x)")

        self.add(ax, labels)

        x_variavel = ValueTracker(0)

        def func(x):
            return np.cos(3*x)
        graph = ax.plot(func, color=PURE_GREEN)

        Ponto_inicial = ax.c2p( x_variavel.get_value(), func( x_variavel.get_value()))
        dot = Dot(point=Ponto_inicial)
        dot.set_color(YELLOW)

        def updater_dot(dot_movel):
            dot_movel.move_to(ax.c2p( x_variavel.get_value(), func( x_variavel.get_value())))

        dot.add_updater(updater_dot)

        self.add(graph, dot)
        self.play(x_variavel.animate.set_value(PI/2), run_time=5)
        dot.remove_updater(updater_dot)
        self.wait()
````

## 5.6.4 Exemplo de `updaters` em gráficos de funções

````python
%%manim -qm Updater_grafico

class Updater_grafico(Scene):
    def construct(self):

        ax = Axes(
            x_range=[0, 3*PI],
            y_range=[-4, 4],
            x_length=6
        ).scale(1)
        ax_labels = ax.get_axis_labels()
        ax_group = VGroup(ax, ax_labels)
        self.add(ax_group)


        #Função com parâmetro k
        def function(x, k):
            return 3.5*np.sin((2+0.1*k)*np.cos(k*x))

        # Rastreador para o parâmetro
        v_k = ValueTracker(0.5)  # Inicializa o rastreador com valor 0.5


        f_1= ax.plot(lambda x: function(x, v_k.get_value()), x_range=[0, 3*PI, 0.01], color=BLUE)
        self.add(f_1)

        # Adiciona o updater à curva
        f_1.add_updater(
            lambda f_1: f_1.become(
                ax.plot(lambda x:  function(x, v_k.get_value()), x_range=[0, 3*PI, 0.01], color=BLUE)
            )
        )

        self.play(v_k.animate.set_value(5), run_time=4, rate_func=linear)

````
