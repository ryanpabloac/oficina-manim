## **5. Manipulando Objetos**
### 5.1 Posições
- `UP` = (0,1,0): 1 unidade acima
- `DOWN` = (0,-1,0): 1 unidade abaixo
- `LEFT` = (-1,0,0): 1 unidade à esquerda
- `RIGHT` = (1,0,0): 1 unidade à direita

As posições podem ser combinadas entre si
- `UL` = (-1,1,0): 1 unidade acima e à esquerda
- `UR` = (1,1,0): 1 unidade acima e à direita
- `DL` = (-1,-1,0): 1 unidade abaixo e à esquerda
- `DR` = (1,-1,0): 1 unidade abaixo e à direita

As posições podem ser combindas com constantes. Ex.: `3*UP + 2*DOWN`

## **6. Planos**

### 6.1 Plano Cartesiano

- Para adicionar somente os eixos cartesianos na tela, basta adicionar a classe `Axes()`
    - Os valores para cada ponto podem ser exibidos utilizando o método `add_coordinates()`;

- Para adicionar os eixos cartesianos com malha quadriculada, basta adicionar a classe `NumberPlane()`
    - `add_coordinates()` serve aqui também;

```python
%%manim -qm -v Warning planoSemGrades

class planoSemGrades(Scene):
    def construct(self):

      eixos=Axes()
      self.play(Create(eixos))
      self.wait(2)

class planoComGrades(Scene):
    def construct(self):

      plano=NumberPlane().add_coordinates()
      self.play(Create(plano))
      self.wait(2)
```
