# 2.2 Criação de Figuras Geométricas

---

As figuras geométricas que apresentaremos serão o triângulo, o quadrado, o retângulo, o círculo e a elipse.

Quando criamos uma figura geométrica no Manim, podemos personalizá-la aplicando diferentes métodos ao objeto. Esses métodos são instruções que modificam a aparência da figura. Vejamos alguns deles e suas funções:

* `color =`: Cor do objeto (dependendo do contexto, refere-se a cor da borda ou do preenchimento);
* `fill_color =`: Cor do preenchimento (caso necessite especificar);

* `stroke_color =`: Cor da borda (caso necessite especificar);

* `fill_opacity =`: Opacidade do preenchimento (de $0$ a $1$);

* `stroke_width =`: Espessura da borda (padrão é $4$);

* `.scale()`: Tamanho do objeto
    * Valores maiores que $1$, aumentam o objeto (por exemplo, ``.scale(2)`` dobra o tamanho original);
    * Valores entre $0$ e $1$, diminuem o objeto (por exemplo, ``.scale(0.5)``, reduz o tamanho original pela metade);
    * Valores negativos, além de redimencionar, eles invertem (espelham) a orientação do objeto em relação ao seu centro (por exemplo, ``.scale(-1)`` mantém o tamanho e inverte a orientação).

* `set_width=`e `set_height=`: Define a largura e altura do objeto.

Podemos utilizar alguns métodos específicos para cada objeto. Por exemplo:

* Quadrado: `side_length =` - Define a medida do lado (se não especificado, o quadrado será criado com lado de medida $1$);
* Retângulo:  `height =` e `width =` - Definem as medidas da altura e largura, respectivamente;
* Círculo: `radium =` - Define o raio  (se não especificado, o círculo será criado com raio de medida $1$);
* Elipse:  `height=` e `width=` - Definem as medidas dos semieixos maior e menor.

## 2.2.2 **Criando Quadrados**

O  `Square()` é usado para criar um quadrado.



```python
# @title
%%manim -qm -v warning Quadrado1

# --- CENA 1: Aparição Simples ---
class Quadrado(Scene):
    def construct(self):

        # --- BASTIDORES (Definição dos Objetos) ---
        # Definindo um quadrado verde com bordas espessas
        Q2 = Square(side_length=4, color=PURE_GREEN, stroke_width=10)

        # --- PALCO (Ação e Apresentação) ---
        self.play(FadeIn(Q2))
        self.wait(5)


# --- CENA 2: Rotação e Transições Suaves ---
class Quadrado1(Scene):
    def construct(self):

        # --- BASTIDORES (Definição dos Objetos) ---
        # Preparando o primeiro quadrado (Vermelho)
        Q1 = Square(side_length=3, color=RED)

        # Preparando o segundo quadrado (Verde Puro) para uso posterior
        Q2 = Square(side_length=4, color=PURE_GREEN)

        # --- PALCO (Ação e Apresentação) ---

        # 1. Entrada do Q1
        self.play(FadeIn(Q1))
        self.wait(1) # pausa

        # 2. Remoção do Q1
        self.remove(Q1)

        # 3. Transição Suave do Q2 (FadeIn e FadeOut)
        # Uso: Para criar efeitos de transição suave de visibilidade.
        self.play(FadeIn(Q2))  # Faz o quadrado aparecer suavemente
        self.wait(1) # pausa
        self.play(FadeOut(Q2)) # Faz o quadrado desaparecer suavemente

        self.wait(1)
```

## 2.2.5 Versão Tracejada (`DashedVMobject()`)

Um **VMobject**  é um "objeto vetorial" padrão do Manim (círculos, quadrados, gráficos de funções), desenhado através de caminhos matemáticos (vetores).

Já o `dashedVMobject()` funciona como um "invólucro" (wrapper) ou uma "capa": ele não cria uma forma nova, mas "envolve" um objeto vetorial já existente e o decompõe em vários pedaços menores.

Alguns argumentos importantes do `dashedVMobject()` são:

    DashedVMobject(

    dashed_ratio=0.6, # Proporção entre traço e espaço (0.6 = 60% traço, 40% vazio)
    dash_offset=0.2   # Deslocamento inicial ("empurra" o início dos traços para o ponto que corresponde 20% do caminho total do objeto)
    )

## 2.3.1 `.move_to` (posição exata)

Usamos `.move_to` para reposicionar um objeto em uma posição específica da cena. Para isso, especificamos as coordenadas do ponto onde o "centro" do objeto será colocado. Algumas posições predefinidas são:

* `.move_to(UP)`: reposiciona para (0, 1, 0), ou seja, 1 unidade para cima;

* `.move_to(DOWN)`: reposiciona para (0, -1, 0), ou seja, 1 unidade para baixo;

* `.move_to(RIGHT)`: reposiciona para (1, 0, 0), ou seja, 1 unidade para a direita;

* `.move_to(LEFT)`: reposiciona para (-1, 0, 0), ou seja, 1 unidade para a esquerda.

Podemos também combinar dois reposicionamentos, por exemplo:

* `.move_to(RIGHT*3+UP*2)`: reposiciona para (3,2,0), ou seja, 3 unidades para a direita e 2 unidades para cima;

* `.move_to(UL)` =`.move_to(UP+LEFT)`: reposiciona para (-1,1,0), ou seja, 1 unidade para a esquerda e 1 unidade para cima;

* `.move_to(DL)` = `.move_to(DOWN+LEFT)`: reposiciona para (-1,-1,0), ou seja, 1 unidade para a esquerda e 1 unidade para baixo;

* `.move_to(DR*2)`= `.move_to(DOWN*2+RIGHT*2)`: reposiciona para (2,-2,0), ou seja, 2 unidades tanto para a direita como para baixo.

```python
%%manim -qm -v warning movendo

# --- CENA 1: Posicionamento Estático ---
class semanimacao(Scene):
    def construct(self):
        # --- BASTIDORES (Definição dos Objetos) ---
        Q1 = Square(side_length=2, color=RED)
        # Q1.move_to(2 * UP)
        # Q1 = Square(side_length=2, color=RED).move_to(2*UP)

        # --- PALCO (Ação e Apresentação) ---
        self.add(Q1)


# --- CENA 2: Movimentação no Plano ---
class movendo(Scene):
    def construct(self):
        # --- BASTIDORES (Definição dos Objetos) ---
        Axes = NumberPlane()
        Q1 = Square(side_length=2, color=RED)
        A = (-4, -2, 0) # Ponto de destino

        # --- PALCO (Ação e Apresentação) ---

        # 1. Criando o plano de coordenadas
        self.play(Create(Axes))

        # 2. Entrada do Quadrado
        self.play(FadeIn(Q1), run_time=3)
        self.wait(2)

        # 3. Movimentação Instantânea (move_to não é animado por padrão)
        # Move o objeto 2 unidades para cima e à esquerda.
        Q1.move_to(UL * 2)
        self.wait(2)

        # 4. Segundo movimento instantâneo
        # Observe que ele considera Q1 na sua posição inicial ainda.
        Q1.move_to(RIGHT * 4)
        self.wait(2)

        # 5. Reposicionando para o ponto específico A
        Q1.move_to(A)
        #Q1.move_to(LEFT*4+DOWN*2)
        self.wait(2)
```

## 2.3.2 `.shift()` (vetor translação)

Usamos `.shift()` para deslocar um objeto a partir da sua posição atual.

A diferença entre `.move_to()` e `.shift()` é que `.move_to()` posiciona o objeto exatamente em um ponto especificado, independentemente de sua posição atual, enquanto `.shift()` apenas desloca o objeto a partir de onde ele já está.

Logo, com uma suave mudança na interpretação, os movimentos predefinidos do Manim para o método `.shift()` são os mesmos usados com `.move_to()`, por exemplo:

  * `.shift(UP)`:  desloca o objeto 1 unidade para cima a partir de onde ele está;

  * `.shift(DOWN)`:  desloca o objeto 1 unidade para baixo a partir de onde ele está;

  * `.shift(RIGHT)`:  desloca o objeto 1 unidade para a direita a partir de onde ele está;

  * `.shift(LEFT)`:   desloca o objeto 1 unidade para a esquerda a partir de onde ele está.


```python
%%manim -qm -v warning Shift

# --- CENA 1: Posicionamento com Shift Estático ---
class semanimacao(Scene):
    def construct(self):

        # --- BASTIDORES (Definição dos Objetos) ---
        # Criando o retângulo e deslocando-o antes de entrar em cena
        R1 = Rectangle(height=3, width=2, color=PINK, fill_opacity=0.7)
        R1.shift(LEFT * 3)

        # --- PALCO (Ação e Apresentação) ---
        self.add(R1)


# --- CENA 2: Diferença entre Move_to e Shift ---
class Shift(Scene):
    def construct(self):
        # --- BASTIDORES (Definição dos Objetos) ---
        # Definindo o retângulo na memória
        R1 = Rectangle(height=3, width=2, color=PINK, fill_opacity=0.7)

        # --- PALCO (Ação e Apresentação) ---

        # 1. Entrada do Retângulo (Correção: run_timer -> run_time)
        self.play(FadeIn(R1), run_time=3)
        self.wait(2)

        # 2. Uso do move_to (Posição absoluta)
        # Move o objeto 2 unidades para cima.
        R1.move_to(UP * 2)
        self.wait(2)

        # 3. Uso do shift (Deslocamento relativo)
        # Vamos usar o shift: desloca o objeto 3 unidades para a esquerda
        # a partir da posição atual (definida pelo move_to anterior).
        R1.shift(LEFT * 3)
        #R1.move_to(LEFT*3)

        self.wait(2)
```

## 2.3.3 `.to_corner()`("quinas" da cena)

É usado para posicionar o objeto em um dos cantos da cena:

* `.to_corner(UL)`: canto superior esquerdo;

* `.to_corner(UR)`:  canto superior direito;

* `.to_corner(DL)`: canto inferior esquerdo;

* `.to_corner(DR)`: canto inferior direito.

```python
%%manim -qm -v warning Mover

# --- CENA 1: Posicionamento nos Cantos Superiores ---
class AN1(Scene):
    def construct(self):
        # --- BASTIDORES (Definição dos Objetos) ---
        # Posicionando o quadrado no canto superior esquerdo e o círculo no direito
        Q1 = Square().to_corner(UL)
        C1 = Circle().to_corner(UR)

        # --- PALCO (Ação e Apresentação) ---
        self.add(Q1)
        self.add(C1)


# --- CENA 2: Movimentação de Cantos para o Centro ---
class Mover(Scene):
    def construct(self):
        # --- BASTIDORES (Definição dos Objetos) ---
        # Definindo as formas nos quatro cantos da tela
        Q1 = Square().to_corner(UL)
        C1 = Circle().to_corner(UR)
        R1 = Rectangle(color=BLUE, fill_opacity=0.9).to_corner(RIGHT + DOWN)
        E1 = Ellipse(height=3, width=2, color=PURE_GREEN, fill_opacity=0.9)
        E1.to_corner(LEFT + DOWN)

        # --- PALCO (Ação e Apresentação) ---

        # 1. Criação simultânea do Retângulo e da Elipse
        self.add(Q1, C1)
        self.wait(2)
        self.play(Create(R1), Create(E1))
        self.wait(2)

        # 2. Movimento instantâneo da Elipse para a origem (centro (0,0,0))
        # move_to usa coordenadas absolutas
        E1.move_to(ORIGIN)

        self.wait(1)
```

## 2.3.4 `.to_edge()` (borda da cena)

Usamos `.to_edge()` para posicionar um objeto junto a uma borda da cena (ao invés de um canto, como faz o `.to_corner()`). Aqui, podemos usar o parâmetro `buff=` (do inglês, buffer) para definir a distância entre o objeto e a borda. Se nada for mencionado, a distância padrão é 0.5.



```python
%%manim -qm -v warning Borda

class Borda(Scene):
    def construct(self):

        # --- BASTIDORES (Definição dos Objetos) ---

        #AX1=NumberPlane().scale(0.5)
        #self.add(AX1)
        # Definindo o círculo com raio, cor de borda, opacidade e preenchimento
        C1 = Circle(
            radius=1.5,
            color=BLUE,
            fill_opacity=0.9,
            fill_color=PURE_GREEN
        )


        # --- PALCO (Ação e Apresentação) ---

        # 1. Criação inicial no centro
        self.play(Create(C1))
        self.wait(2)

        # 2. Movimento para a borda superior com respiro (buff) de 1 unidade
        C1.to_edge(UP, buff=1)
        self.wait(2)

        # 3. Salto para a borda inferior (buff padrão de 0.25)
        C1.to_edge(DOWN)
        self.wait(2)
        self.play(Create(C1))
        self.wait(2)

        # 4. Retorno para a borda superior (sem o buff extra de 1)
        C1.to_edge(UP)
        self.wait(2)

        # 5. Movimento final para a borda direita
        C1.to_edge(RIGHT)
        self.wait(2)
```

## 2.3.6 `.copy()` (cópia idêntica)

Para criar cópias independentes de objetos, utilizamos o método `.copy()`.


```python
%%manim -qm -v warning Copia

# --- CENA 1: Adição Instantânea de Cópias ---
class semanimacao(Scene):
    def construct(self):

        # --- BASTIDORES (Definição dos Objetos) ---
        # Criando o retângulo original
        R1 = Rectangle(height=2, width=3, color=BLUE, fill_opacity=0.7)

        # Criando o Clone (R2) e enviando para a borda direita
        R2 = R1.copy().to_edge(RIGHT)

        # --- PALCO (Ação e Apresentação) ---
        self.add(R1)
        self.wait(2)

        self.add(R2)
        self.wait(2)


# --- CENA 2: Animação com Clone ---
class Copia(Scene):
    def construct(self):
        # --- BASTIDORES (Definição dos Objetos) ---
        # Definindo R1 posicionado à esquerda
        R1 = Rectangle(height=2, width=3, color=BLUE, fill_opacity=0.7)
        R1.move_to(LEFT * 3)

        # --- PALCO (Ação e Apresentação) ---
        # 1. Entrada suave do primeiro retângulo
        self.play(FadeIn(R1))

        # 2. Criando o Clone e animando sua aparição na borda oposta
        # Note que o copy() preserva cor e opacidade, mas to_edge define o novo lugar
        R2 = R1.copy().to_edge(RIGHT)

        self.play(Create(R2))

        self.wait(2)
```

# 2.4 Animações utilizando `.animate`

O `.animate` é um modificador poderoso que transforma métodos estáticos em animações fluidas. Ele atua como uma ponte: avisa ao Manim que a próxima mudança (como mover, colorir ou girar) deve ser "filmada" passo a passo. Por isso, ele nunca funciona sozinho; deve ser sempre utilizado dentro de um `self.play()`. Uma de suas maiores vantagens é permitir o encadeamento de várias transformações em uma única linha de código.

```python
%%manim -qm variosmetodos

class variosmetodos(Scene):
    def construct(self):
        Q1=Square(side_length=3,color=RED,stroke_color=YELLOW_C,fill_opacity=0.8)

        self.add(Q1)
        self.wait(3)
        self.play(Q1.animate.scale(1.5).shift(RIGHT).set_color(BLUE).rotate(PI/3))
        self.wait(1)


class Quadrado(Scene):
    def construct(self):
        Q1=Square(side_length=3,color=RED)

        self.add(Q1)
        self.play(Q1.animate.rotate(PI/3))
        self.wait(1)

        self.remove(Q1)

        Q2 = Square(side_length=4,color=PURE_GREEN)
        self.play(FadeIn(Q2))
        self.wait(1)
        self.play(FadeOut(Q2))

        self.wait(1)

class D1(Scene):
    def construct(self):
        left_Q1 = Square(color=BLUE, fill_opacity=0.7).shift(2*LEFT)
        right_Q2 = Square(color=GREEN, fill_opacity=0.7).shift(2*RIGHT)

        self.play(Rotate(right_Q2, angle=PI), run_time=3)

        self.play(left_Q1.animate.rotate(PI/2), run_time=3)

        #Diferentemente do Rotate(), aqui não há necessidade de especificar explicitamente
        #que o objeto será animado; o .animate já assume que a transformação será animada.
        self.wait()
```

# 2.5 Animação de Transformação entre Objetos

O Manim oferece diversos tipos de transformações, todas baseadas no método `Transform`, mas adaptadas para diferentes propósitos e efeitos visuais. Vejamos os principais tipos de transformações:

* `Transform()`: Transforma um objeto em outro de maneira suave. Tecnicamente, o primeiro objeto não é substituído: ele assimila as propriedades geométricas e visuais do segundo. Na prática, o primeiro objeto permanece na cena, mas "fantasiado" do segundo, mantendo seu nome original nos bastidores do código.

* `ReplacementTransform()`: Visualmente igual ao `Transform`, porém remove o objeto original ao final da transformação. Utilizado quando se deseja que o original "desapareça" no processo.

* `TransformFromCopy()`: Cria uma cópia do primeiro objeto, que se transforma no segundo objeto, enquanto o primeiro permanece intacto.

* `TransformMatchingShapes()`: Tenta mapear as partes semelhantes da geometria entre dois objetos para que a transição pareça mais natural e contínua. É especialmente útil quando se deseja transformar uma forma em outra mantendo uma correspondência visual entre contornos ou áreas semelhantes.

* `TransformMatchingTex()`: Transforma uma expressão matemática em outra, alinhando automaticamente os símbolos iguais para que a transição fique mais suave.




```python
%%manim -qm -v warning Transform1

# --- CENA 1: Transformações com Herança de Objeto ---
class Transform1(Scene):
    def construct(self):
        # --- BASTIDORES ---
        Q = Square(color=PURE_GREEN, stroke_color=BLUE, stroke_width=6).shift(DOWN * 1.5)
        T = Triangle(color=BLUE, fill_opacity=1, stroke_width=4, stroke_color=RED).shift(UP * 2)
        R = Rectangle(color=PURE_GREEN, fill_opacity=0.5, height=3, width=5, stroke_color=RED)

        # --- PALCO ---
        self.add(Q)
        self.wait(2)

        # O Q se transforma visualmente em T, mas quem permanece é o Q (fantasiado de T)
        self.play(Transform(Q, T))
        self.wait(2)
        #Mostrar que T não está em cena
        #self.play(Transform(T,R))
        #self.play(Transform(Q,R))
        self.wait(2)

        # Transforma o quadrado em um triângulo a partir de uma cópia
        #self.play(TransformFromCopy(Q, T))
        self.wait()


# --- CENA 2: Transformações de Fórmulas (Tex) ---
class Transform2(Scene):
    def construct(self):
        # --- BASTIDORES ---
        self.camera.background_color = WHITE

        # Usando {{ }} para isolar as partes que o Manim deve rastrear e mover
        eq1 = MathTex("42 {{ a^2 }} + {{ b^2 }} = {{ c^2 }}", color=RED)
        eq2 = MathTex("42 {{ a^2 }} = {{ c^2 }} - {{ b^2 }}", color=PURE_GREEN)
        eq3 = MathTex(r"a^2 = \frac{c^2 - b^2}{42}", color=PURE_BLUE)

        # --- PALCO ---
        self.add(eq1)
        self.wait()

        # Move os termos da equação de forma inteligente
        self.play(TransformMatchingTex(eq1, eq2))
        self.wait()

        # Transforma as formas dos caracteres de uma equação na outra
        self.play(TransformMatchingShapes(eq2, eq3))
        self.wait()


# --- CENA 3: Substituição Completa ---
class Transform3(Scene):
    def construct(self):
        # --- BASTIDORES ---
        C1 = Circle(radius=2, color=RED)
        R1 = Rectangle(color=BLUE_C, fill_opacity=0.5, height=3, width=5, stroke_color=RED)
        T = Triangle(color=BLUE, fill_opacity=1, stroke_width=4, stroke_color=RED).shift(UP * 2)

        # --- PALCO ---
        self.add(C1)
        self.wait(2)

        # Substitui completamente C1 por R1.
        # C1 é removido da memória e R1 assume o posto oficialmente.
        self.play(ReplacementTransform(C1, R1))
        self.wait(2)
```

## 2.5.2 ``MoveToTarget()``

É uma animação usada para mover um objeto (Mobject) de sua posição atual para uma posição/estado “alvo” (target) que você define previamente.


Passo a passo de como usá-lo:

1. Criar o objeto:
    * ``M1=Circle()`` ou Square, Triangle, etc

2. Gerar o target:
    * ``M1.generate_target()``

3. Modificar o target:
    * `` M1.target.shift(RIGHT * 3)``
    *  ``M1.target.set_color(RED)``
    * `` M1.target.scale(1.5)``

4. Aplicar ``MoveToTarget()`` com ``self.play()``:
    * ``self.play(MoveToTarget(M1))``



O ``MoveToTarget()`` é útil quando:

* Você quer preparar o estado final com antecedência;
* Precisa fazer várias modificações mais complexas.



```python
%%manim -qm -v warning Exemplo1

#from manim import * # Sem isso ele estava dando erro.

class Exemplo1(Scene):
    def construct(self):
        M1 = Square(color=BLUE,fill_color=YELLOW_C,fill_opacity=1).scale(1.2)

        self.play(Create(M1))
        self.wait()

        # Cria o target
        M1.generate_target()

        # Modifica o target
        M1.target.shift(RIGHT * 3)
        M1.target.set_color(RED)
        M1.target.scale(1.5)
        M1.target.set_fill(WHITE, opacity=1)
        M1.target.rotate(PI/3)

        # Anima até o target
        self.play(MoveToTarget(M1))

        self.wait()
```
