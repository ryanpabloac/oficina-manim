# 3.2 Alguns Tipos de Planos


A escolha do plano no Manim define a "gramática" visual da sua animação, permitindo que o ambiente se ajuste à natureza do problema matemático proposto.

O **Plano Cartesiano** organiza o espaço em eixos ortogonais $(x, y)$.

O **Plano Complexo**, embora visualmente similar ao cartesiano, possui métodos nativos (como o `n2p()`) que permitem tratar pontos diretamente como números complexos $a + bi$.

Já o **Plano Polar**  é voltado a simetrias circulares, facilitando a plotagem baseada em raio ($r$) e ângulo ($\theta$).



## 3.2.1 Plano Cartesiano

Para adicionar o Plano Cartesiano podemos usar a classe `Axes()` (apenas com os eixos coordenados) e o `NumberPlane()`(além dos eixos, com a malha quadriculada).

Agora, se desejarmos que os valores numéricos (rótulos) apareçam ao longo dos eixos é necessário aplicar o método `.add_coordinates()`.


````python
%%manim -qm -v Warning plano1

class plano1(Scene):
    def construct(self):

      eixos=Axes()
      self.play(Create(eixos))
      self.wait(2)


class plano2(Scene):
    def construct(self):

      eixos=Axes()
      self.play(Create(eixos))
      self.wait(2)

      ponto = Dot([2,1,0], color=YELLOW)

      ponto = Dot(eixos.c2p(2,1), color=YELLOW) # Indexa o ponto à coordenada (2,1) do sistema de coordenadas Ax

      self.play(Create(ponto))
      self.wait(4)

class plano3(Scene):
    def construct(self):

      eixos=Axes()
      ponto = Dot(eixos.c2p(2,1), color=YELLOW) # Indexa o ponto à coordenada (2,1) do sistema de coordenadas Ax

      self.play(Create(eixos))

      #self.play(Create(ponto)) ou então:
      eixos.add(ponto) # Adiciona o ponto ao objeto eixos
      self.wait(3)


      self.play(eixos.animate.shift(UR))
      self.wait(5)



class plano4(Scene):
    def construct(self):

      eixos=NumberPlane().add_coordinates() #com malha e rótulos para os números nos eixos
      self.play(Create(eixos))
      self.wait(2)


      # Para definir vetor basta informar as componentes do vetor.
      ponto = Dot([2,1,0]).set_color(YELLOW)
      vetor = Vector([-5,-2,0], color=GREEN) # Renderiza o vetor com origem em (0,0) do sistema de coordenadas padrão do Manim.

      self.play(GrowArrow(vetor)) #"GrowArrow" é uma animação específica para vetores.
      self.play(Create(ponto))
      self.wait(2)
````

#3.3 Explorando o Plano Cartesiano $xy$

Agora que já sabemos criar os eixos coordenados, vamos personalizá-los. Algumas ações que podemos fazer são:

*  `x_range=[a, b, dx]`: define o intervalo do eixo x, de $a$ até $b$, onde $dx$ é o espaçamento (incremento) entre as coordenadas.

*  `y_range=[c, d, dy]`: define o intervalo do eixo $y$, de $c$ até $d$, onde $dy$ é o espaçamento entre as coordenadas.

* `x_length=n`: define o comprimento visual do eixo $x$ na tela, que terá n unidades de comprimento na tela, independente dos valores de $a$ e $b$.

* `y_length=m`: define o comprimento visual do eixo $y$ na tela, que terá m unidades de comprimento na tela, independente dos valores de $c$ e $d$.

* `axis_config={}`: permite definir configurações globais para ambos os eixos simultaneamente. Abaixo estão os principais argumentos que você pode utilizar dentro das chaves:

  * `"color": ` define a cor dos eixos e das marcações (a cor padrão no Manim é branca);

  * `"stroke_width": ` define a espessura dos eixos (a espessura padrão no Manim é 2);

  * `"include_tip":`(True ou False): define se os eixos terão setas (o padrão no Manim é ter setas);
  
  * `"font_size":` ajuste do tamanho da fonte dos números (rótulos) nos eixos (o tamanho padrão da fonte no Manim é 36);

  * `"line_to_number_buff"`: define a distância (espaçamento) entre o eixo e o número (o espaçamento padrão no Manim é 0.25).


````python
%%manim -qm Ax2

class Ax1(Scene):
    def construct(self):
        # Fundo
        self.camera.background_color = WHITE
        # Configurar eixos
        axes1 = Axes(
            x_range=[-10, 10,1], # Intervalo do eixo x (mínimo, máximo, passo)
            y_range=[-6, 6, 1], # Intervalo do eixo y (mínimo, máximo, passo)
            x_length=10, # Comprimento do eixo X
            y_length=6, # Comprimento do eixo Y
            axis_config={"color": PURE_GREEN},  # Cor dos eixos
            ).scale(1.2)

        axes1.add_coordinates()

        # Adicionar rótulos aos eixos
        labels1= axes1.get_axis_labels(x_label="x", y_label="f(x)")
        labels1.set_color(RED)

        self.play(Create(axes1))
        self.play(Write(labels1))
        self.wait(2)


class Ax2(Scene):
    def construct(self):
        axes2 = Axes() # Axes é um VGroup (um grupo de objetos visuais) que contém 2 elementos: [0] a linha do x e [1] a linha do y
        eixo_x = axes2.get_axes()[0].set_color(BLUE) # Adiciona eixo x
        eixo_y = axes2.get_axes()[1].set_color(RED)  # Adiciona eixo y

        self.add(eixo_x,eixo_y)
````

## 3.3.2 Criação de gráfico de função $y=f(x)$ no plano $xOy$

Para a construção de gráficos, utilizaremos a biblioteca NumPy (Numerical Python), comumente abreviada como `np`. Ela é a base da computação científica em Python, permitindo manipular grandes volumes de dados com alta performance. Diferente das listas comuns, o NumPy utiliza vetorização, processando arrays inteiros simultaneamente sem loops explícitos (como o `for`).

No Manim, os gráficos são gerados a partir de funções Python estruturadas da seguinte forma:


```
 def nome_da_funcao(x):
     return lei_da_funcao
```

 Veja o exemplo abaixo utilizando as funções:
 `np.cos(x)`, `np.sin(x)`, `x**2`, `np.exp(x)`, np.log(x)



````python
# @title
%%manim -qm Axes2

class Axes2(Scene):
    def construct(self):
        # Fundo

        # Configurar eixos
        AX1 = Axes(
            x_range=[-5, 5,1], # Intervalo do eixo x (mínimo, máximo, passo)
            y_range=[-4, 4, 1], # Intervalo do eixo y (mínimo, máximo, passo)
            x_length=10, # Comprimento do eixo X
            y_length=8, # Comprimento do eixo Y
            axis_config={"color": WHITE},  # Cor dos eixos
            ).scale(1).add_coordinates()
        #Adicionar cor aos eixos
        #axes1.set_color(PURE_GREEN)
        # Adicionar rótulos aos eixos
        labels1= AX1.get_axis_labels(x_label="x", y_label="y")
        labels1.set_color(WHITE)

        self.add(AX1,labels1)
        self.wait(1)

        #Definir funções funções y=f(x)

        def f1(x):
          return x**2   # Usando np: np.power(x, 2) - Para função exponenciais quaisquer, np.power(base, expoente)

        def f2(x):
          return np.sin(x) # Para funções predefinidas usa np.f(x)

        def f3(x):
          return np.exp(x)

        #Outra maneira de definir y=f(x)

        f4=lambda x: np.cos(2*x)

        #Grafico de y=f(x) quando já definimos f(x) antes

        grafico_f1=AX1.plot(f1,color=YELLOW_C)


        #Grafico de y=f(x) definindo f(x) dentro da função plot
        grafico_f2=AX1.plot(lambda x: np.sin(x),color=RED)

        #Restrição do intervalo do x no gráfico f3
        grafico_f3=AX1.plot(f3,color=BLUE, x_range=(-3, 3, 0.001))

        #Espessura da linha

        #grafico_f1.set_stroke_width(5)

        # Transparência

        #grafico_f1.set_opacity(0.7)

        # Gráficio tracejado

        Gf4=AX1.plot(f4,color=GREEN)

        grafico_tracejado=DashedVMobject(
            Gf4,
            num_dashes=100,    # Quantidade de "tracinhos"
            dashed_ratio=0.5, # Proporção entre traço e espaço (0.7 = 70% traço, 30% vazio)

            )

        self.play(Create(grafico_tracejado))

        # Animação

        self.play(Create(grafico_f1))
        self.wait(2)
        self.play(Create(grafico_f2))
        self.wait(2)
        self.play(Create(grafico_f3))

        self.wait(2)


````

## 3.3.3 Adicionando Rótulos a Gráficos

Para tornar seus gráficos mais compreensíveis, é essencial adicionar rótulos (label) que identifiquem as funções. O Manim facilita isso com o método `get_graph_label()`. É frequentemente usado após ter criado um gráfico com `plot()`.

A sintaxe básica para usar o método é:

```
rotulo = AX1.get_graph_label(nome_funcao, rótulo, x_val, direction, color)
```

Vamos entender cada um dos parâmetros:

* `AX1`: "nome" do sistema de coordenadas já definido.
* `nome_funcao`: função já definica (geralmente criada com o método `plot()`) cujo gráfico você quer adicionar um rótulo.
* `rótulo`: texto que será exibido como rótulo. Você pode usar formatação LaTeX para equações complexas, como `r"$f(x) = \sin(x)$"`.
* `x_val`: Define a posição horizontal do rótulo, ou seja, o valor de x no qual ele será colocado.
* `direction`: Controla a direção em que o rótulo será posicionado em relação ao ponto no gráfico. Use direções como `UP`, `DOWN`, `LEFT` ou `RIGHT`.
* `color`: Define a cor do rótulo. Se não for especificada, a cor do próprio gráfico é usada por padrão.

Após criar o rótulo, você pode adicioná-lo à sua cena. Por exemplo, você pode usar o método `add()` para incluí-lo na animação.


````python
# @title
%%manim -qm Rotulo

class Rotulo(Scene):
    def construct(self):
        ax = Axes(
            x_range=[-4, 4, 1],
            y_range=[-1.5, 1.5, 1],
            x_length=8
        ).add_coordinates()

        f1 = ax.plot(lambda x: np.sin(x), x_range=[-4, 4, 0.01], color=GREEN)
        f1_label = ax.get_graph_label(f1, label=MathTex(r"y=\sin(x)"), x_val=2, direction=UP*1.5,color=GREEN)
        f1_group = VGroup(f1, f1_label)

        f2 = ax.plot(lambda x: np.cos(x), x_range=[-4,4, 0.01], color=PINK)
        f2_label = ax.get_graph_label(f2, MathTex(r"y=\cos(x)"), x_val=2.5, direction=DOWN*2.5)
        f2_group = VGroup(f2, f2_label)

        self.add(ax)
        self.play(Create(f1_group))
        self.play(Create(f2_group))
        self.wait(2)
````

# 3.4 Adicionado gráficos no plano Polar

No sistema de coordenadas polares, a posição de um ponto não é definida por $(x, y)$, mas sim pelo par $(r, \theta)$, onde:

* $\theta$ (theta): Representa o ângulo (geralmente medido em radianos);

* $r$: Representa a distância radial do ponto até a origem.

Uma função polar é expressa na forma:$$r = f(\theta).$$

💡
1. Para plotar esses pontos corretamente, utilizamos a função `AX1.plot_polar_graph(
    r,
    theta_range=[inicio, fim]
)`
2. Também podemos usar `polar_to_point(r, theta)`. Ela realiza a conversão automática para o sistema de coordenadas cartesianas da tela, respeitando fielmente a escala do seu plano.



````python
# @title
%%manim -qm graficop

class graficop(Scene):
    def construct(self):
      Ax2=PolarPlane(
            radius_max=4,
            azimuth_units="PI radians",
        ).add_coordinates()
      self.play(Create(Ax2))
      self.wait(2)
      #Definir a Função
      def r1(t,k):
            return 3

      def r2(t,k):
            return k*np.cos(k*t)


      def r3(x,k):
            return 2*np.sin(1.2*k*x)**2+2*np.cos(k*x)**3

      def r4(x,k):
            return 1.8*np.cosh(np.cos(2*k*x)*np.sin(k*x))

      def r5(x, k):
            return 3*np.sin(k*np.cos(k*x))

     #usado a varipavel theta plotar a função $r=f(theta)$

      def r(theta):
        return  3 * np.sin(4 * theta)


      curva1 = Ax2.plot_polar_graph(
            r    ,
            theta_range=[0, 2*TAU],
            color=PURE_YELLOW
        )
      # Outra opção
      curve2 = ParametricFunction(
            lambda t: Ax2.polar_to_point(r2(t,2), t),
            t_range=[0, 12 * PI], # Variação do t
            color=PURE_GREEN # cor do gráfico
            )

      self.play(Create(curva1),run_time=5)
      self.wait(2)
      #self.play(ReplacementTransform(grafico_polar,curve1),run_time=5)
      #self.wait(2)
````

# 3.5 Espaço 3D

O Manim permite criar animações em três dimensões, oferecendo mais realismo e profundidade às representações matemáticas. Para gerar os eixos coordenados, utilizamos o objeto `ThreeDAxes()`.

No espaço 3D, podemos posicionar objetos em diferentes planos e rotacionar a cena para observá-los de vários ângulos. Para controlar a perspectiva da câmera, utilizamos os ângulos $\phi$ e $\theta$:

* $\phi$ (phi): define o ângulo vertical da visão (inclinação da câmera);

* $\theta$ (theta): define o ângulo horizontal (giro ao redor do eixo vertical $z$).


````python
# @title
%%manim -qm plano5

class plano5(ThreeDScene):  # Use ThreeDScene
    def construct(self):
        Ax3 = ThreeDAxes().add_coordinates()
        self.play(Create(Ax3))
        self.move_camera(phi=65 * DEGREES, theta=60 * DEGREES) # a perspectiva da câmera em relação aos objetos 3D na cena
        self.wait(2)

        #Algumas superfícies

        cone=Cone()
        Toro=Torus()
        esfera=Sphere()
        cilindro=Cylinder()
        cubo=Cube()
        Prisma=Prism()

        # Usamos mais superfícies paramétricas

        esfera = Surface(
            lambda u, v: np.array([
                2 * np.cos(v) * np.cos(u),
                2 * np.sin(v) * np.cos(u),
                2 * np.sin(u)]),
            u_range=[-PI, PI],
            v_range=[0, 2 * PI],
            resolution=(15, 32),
            )
        esfera.set_fill_by_checkerboard(PURE_GREEN, BLUE,ORANGE,PINK, opacity=0.5)

        paraboloide=Surface(
            lambda u, v: np.array([
                np.cos(v)*u,
                np.sin(v)*u,
                u**2
            ]),v_range=[0, 2 * PI],
            u_range=[0, 1.7],
            resolution=(15, 32) ).scale(0.7)

        paraboloide.set_fill_by_checkerboard(PINK, BLUE, opacity=0.5)

        toro = Surface(
            lambda u, v: np.array([
                (3 + 1 * np.cos(v)) * np.cos(u),
                (3 + 1 * np.cos(v)) * np.sin(u),
                1 * np.sin(v)]),
            v_range=[0, 2 * PI],
            u_range=[-PI, PI],
            resolution=(15, 32)).scale(0.6)

        toro.set_fill_by_checkerboard(RED, BLUE, YELLOW, opacity=0.5)

        catenoide = Surface(
            lambda u, v: np.array([
                0.5 * np.cosh(v / 0.5) * np.cos(u),
                0.5 * np.cosh(v / 0.5) * np.sin(u),
                v]),
            v_range=[-1.2, 1.2],
            u_range=[-PI, PI],
            resolution=(5, 32)).scale(0.7)

        tube2 =Surface(
            lambda u, v: np.array([
                np.sin(v),
                np.cos(u)*np.sin(2*v),
                np.sin(u)*np.sin(2*v)
            ]), u_range=[0, 2*PI], v_range=[0, 2*PI]).scale(1.3)


        tube2.set_fill_by_checkerboard(GREY_BROWN,RED, opacity=0.5)

        self.play(Create(esfera),run_time=7)
        self.wait(2)
````
