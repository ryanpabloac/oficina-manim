# 4.1 Como definir uma função?

No Python e no Manim, existem duas formas principais de definir funções matemáticas:

* Usando a palavra-chave `def`, por exemplo,

```
def nome_da_função(x): # x é a variável independente
    y = x**2 - 4
    return y
```

🔎 Quando você usa as funções tradicionais (seno, cosseno, polinomiais, etc) é preciso chamar um método (`.plot()`ou `get_graph()`) para gerá-las, por exemplo:

```
grafico=ax.plot(nome_da_função,
                x_range=[a,b],
                color=PURE_GREEN)

```

* Usando `lambda`.

As funções `lambda` são funções rápidas que não precisam, necessariamente, de um nome e são escritas em uma única linha de código, direto no gráfico da função, por exemplo,

```
grafico = ax.plot(lambda x: x**2 - 4, x_range=[-2, 2])
```

🔎 Observação: É claro que, se conveniente, podemos nomear a função mesmo utilizando `lambda`:

```
nome_da_funcao = lambda x: x**2-4
grafico = ax.plot(nome_da_funcao, x_range=[-2, 2])
```


# 4.2 Gráfico de Função Definida por Partes

Recomendamos criar mais de uma função:

$f\left(  x\right)  =\left\{
\begin{array}
	[c]{c}%
	f_1(x);\text{ se }a<x \leq b\\
	f_2(x);\text{ se }b< x\leq c \\
  f_3(x);\text{ se }c< x\leq d \\
	f_4(x);\text{ se }x> e\\
\end{array}
\right.$


````python
# @title
%%manim -qm Funcao_partes

class Funcao_partes(Scene):
    def construct(self):
      ax=Axes().add_coordinates()
      x_label=ax.get_axis_labels(x_label="x", y_label="y")
      self.add(ax,x_label)


     # Intervalos do domínio
      a = -7
      b = 0
      c =2
      d = 4
      e = 7     # ou então:  a, b, c, d, e = -7, 0, 2, 4, 7

      # Definindo as leis das funções
      f1=lambda x: np.cos(x)
      f2=lambda x: np.abs(x) # função módulo (valor absoluto)
      f3=lambda x: np.sin(x)
      f4=lambda x: 2

      grafico_f1=ax.plot(f1,x_range=[a,b],color=PURE_GREEN)
      grafico_f2=ax.plot(f2,x_range=[b,c],color=YELLOW_C)
      grafico_f3=ax.plot(f3,x_range=[c,d],color=RED)
      grafico_f4=ax.plot(f4,x_range=[d,e],color=BLUE_C)

      # Delimitando os extremos dos intervalos no domínio
      c_f1a=Circle(radius=0.07,color=PURE_GREEN).move_to(ax.c2p(a,f1(a)))
      c_f1b=Circle(radius=0.07,color=PURE_GREEN,fill_color=PURE_GREEN,fill_opacity=1).move_to(ax.c2p(b,f1(b)))

      # Mão no código: represente no gráfico os demais extremos.

      self.play(Create(c_f1a),Create(grafico_f1),Create(c_f1b),run_time=5)
      self.wait(2)
      self.play(TransformFromCopy(grafico_f1,grafico_f2)) # Se usar apenas Transform(), o f1 deixa de existir e se transforma em f2
      self.wait(2)
      self.play(TransformFromCopy(grafico_f2,grafico_f3))
      self.wait(2)
      self.play(TransformFromCopy(grafico_f3,grafico_f4))

      self.wait(2)
````


# 4.3 Gráfico de Função Implícita: interpretada como a curva de nível $f(x,y)=0$

A classe que utilizamos para representar uma função implícita é o `ImplicitFunction()`.

Para garantir que a função implícita seja desenhada corretamente de acordo com a escala do plano no Manim, devemos utilizar  o atributo `min_depth`.


````python
# @title
%%manim -qm fimplicita

class fimplicita(Scene):
    def construct(self):
        # Fundo

        # Configurar eixos
        axes1 = Axes(
            x_range=[-3, 3, 1],
            y_range=[-3, 3, 1],
            axis_config={"color": WHITE},
            ).scale(1)

        # Definindo os rótulos dos eixos
        labels1= axes1.get_axis_labels(x_label="x", y_label="y")
        labels1.set_color(WHITE)

        self.add(axes1,labels1)

        # Definir f(x,y)

        f_xy= lambda x, y: (x**2 + y**2 - 1)**2 - x**2 * y**2


        curva1 = axes1.plot_implicit_curve(
            f_xy,
            color=PURE_GREEN,

        )

        # Outra modo de fazer o gráficos implícitos mais fácil

        # Função implícita da elipse: x²/a^2 + y²/b^2 - 1
        curva2 = ImplicitFunction(
               lambda x, y: x**2+(y**2)/4 -1,
               color=BLUE,
               )
        self.play(Create(curva1))
        self.wait(1)

        self.play(ReplacementTransform(curva1,curva2))
        self.wait(1)

        self.remove(curva2)

        graph1 = ImplicitFunction(
            lambda x, y: (x**2+y**2-1)**3-(x**2)*(y**3),
            color=RED
        )

        self.play(Create(graph1))

        self.wait(2)
````

# 4.4 Gráfico de Funções Paramétricas: $r(t)=(x(t),y(t),z(t))$

 A classe `ParametricFunction()` no Manim permite criar curvas paramétricas: ela aceita uma função $r(t)$ que retorna as coordenadas $(x(t),y(t))$ para curvas 2D ou $(x(t),y(t),z(t))$ para curvas 3D, onde o parâmetro $t$ varia dentro de um intervalo especificado.
Por exemplo:


```
ParametricFunction(lambda t: np.array($[x(t),y(t),z(t)]$),
    t_range=[t_min, t_max],  # Intervalo do parâmetro t
    color=WHITE, # Cor da curva
    stroke_width=2, # Espessura da linha
)
```
🔎 A classe `ParametricFunction( )` já é, por si só, um objeto geométrico (um VMobject), ou seja, ela já nasce como gráfico pronto. Isso significa que não precisa chamar o método `.plot()`, por exemplo.


📝 Lembre-se:

A melhor abordagem para garantir que a curva paramétrica siga corretamente a escala dos `ax=Axes()` definido é utilizar `ax.c2p(x, y)`, que converte as coordenadas do gráfico para o sistema de coordenadas `ax`.




````python
# @title
%%manim -qm parametrica

class parametrica(Scene):
    def construct(self):
        ax = Axes(
            x_range=[-3, 3],
            y_range=[-3, 3],
            x_length=6,
            y_length=6
        ).add_coordinates()
        ax.scale(1) #Alterar a escala
        ax_labels = ax.get_axis_labels()
        ax_group = VGroup(ax, ax_labels)

        self.add(ax_group)

        circulo_g = ParametricFunction(
            lambda t: ax.c2p(2 * np.cos(t),
                2 * np.sin(t)),
            color=BLUE,
            t_range=[0, 2 * PI, 0.01], # Definição do intervalo de t
            stroke_width=5, # Espessura da linha
        )
        E_circulo_g=Tex(r"$r_1(t)=(2\cos(t),2\sin(t))$",color=RED).next_to(ax,DOWN*0.7)

        grupo1=VGroup(E_circulo_g,circulo_g)
        self.play(
            Create(grupo1, run_time=2),
        )


        epicicloide = ParametricFunction(
            lambda u: np.array([
                1.5 * np.cos(u) - 0.5 * np.cos((5 * u)),
               1.5 * np.sin(u) - 0.5 * np.sin((5 * u)),
                0
            ]),
            color=PURE_GREEN, t_range=np.array([0, 4 * PI, 0.01])
        ).move_to(ax.c2p(0, 0))

        E_epicicloide=Tex(r"$r_4(t)=(1.5\cos(t)-0.5\cos(5t),1.5\sin(t)-0.5\sin(5t))$",color=PURE_GREEN).next_to(ax,DOWN*0.8).scale(0.7)
        etiqueta_epicicloide=Text("Epicicloide",color=RED).move_to(DOWN * 2.5)

        epicicloide_grupo=VGroup(epicicloide,etiqueta_epicicloide,E_epicicloide)
        self.wait(2)


        self.play(ReplacementTransform(grupo1,epicicloide_grupo))

        self.wait(2)
````

# 4.5 Adicionando Pontos e Linhas no Plano Cartesiano $xOy$

* **Pontos no Plano Cartesiano $xOy$:**

Para adicionar pontos ao plano cartesiano usando a biblioteca Manim, podemos usar a classe `Dot()`.

Para acessar as coordenadas desse ponto usamos os métodos `get_x()` (abscissa do ponto) e `get_y()` (ordenada do ponto).

* **Linhas no Plano Cartesiano xOy:**

A classe `Line( )` é a principal ferramenta para desenhar linhas retas. Para criá-la, precisamos de um ponto de início e um ponto de fim:

`L1= Line(Ponto_inicial_linha, Ponto_final_linha)`.

Uma vez que a linha é criada, podemos acessar suas propriedades e manipulá-las:

- Ponto Inicial: `L1.start`

- Ponto Final: `L1.end`

- Ponto Médio: `L1.get_center()`

- Comprimento: `L1.get_length()`.


Para criar uma linha tracejada, usamos a classe `DashedLine()`, como já visto na Oficina 2.

````python
# @title
%%manim -qm PL1

class PL1(Scene):
    def construct(self):

        #self.camera.background_color =WHITE

        # Configurar eixos
        axes1 = Axes(
            x_range=[-4, 4, 1],
            y_range=[-4, 4, 1],
            x_length=8,
            y_length=8,
            axis_config={"color": WHITE},
          ).add_coordinates()
        #axes1.set_color(PURE_GREEN)

        # Adicionar rótulos aos eixos
        labels1= axes1.get_axis_labels(x_label="x", y_label="y")
        labels1.set_color(WHITE)

        # VGgroup cria um grupo de objetos

        grupo_axes=VGroup(axes1,labels1)

        self.add(grupo_axes)

        #Criando pontos

        P1=Dot(axes1.c2p(0,0),radius=0.05,color=RED)
        P2=Dot(axes1.c2p(3,2),radius=0.05,color=BLUE_C)

        # Outra maneira de adicionar pontos com centro fora da origem do sistema padrão do Manim:
        P3=Dot(color=PURE_GREEN).move_to(axes1.c2p(2,0))
        P4=Circle(radius=0.07,color=YELLOW_C).move_to(axes1.c2p(4,3))
        P5=Dot(axes1.c2p(3,3),color=PURPLE_C)
        self.add(P1)
        self.add(P2)
        self.add(P3,P4,P5)

        text_P5=Tex(r"P5",color=PURPLE_C).next_to(P5,UP*0.5).scale(0.8)
        self.add(text_P5)

        # Mão no código: acrescente legendas para os demais pontos.

        # Definindo uma linha:
        L1=Line(P1,P2).set_color(YELLOW_C)
        self.add(L1)

        comprimento=L1.get_length()
        texto_comprimento=MathTex(rf"C= {comprimento:.2f}", color=YELLOW).next_to(L1, UP*0.2).scale(0.7)


        self.play(Create(L1))
        self.play(Write(texto_comprimento))
        self.wait(2)

        #Usando ponto final da linha L1

        L2=DashedLine(L1.end,P5).set_color(RED)
        self.play(Create(L2))
        self.wait(2)

        #Rot=L2.rotate(PI/2,about_point=L2.get_center())
        #self.play(Create(Rot))
        #self.wait(2)








````

# 4.7 Adicionando pontos sobre o gráfico de uma função

 É importante destacar a diferença entre **adicionar pontos diretamente no plano cartesiano** e **adicionar pontos no gráfico de uma função**:  

- Quando colocamos um ponto no **plano cartesiano**, usamos suas **coordenadas $(x,y)$** de forma direta.  
- Já quando colocamos um ponto no **gráfico de uma função**, o valor de $y$ não é escolhido livremente; ele é obtido a partir da função, isto é, $y = f(x)$.

````python
# @title
%%manim -qm PL

class PL(Scene):
    def construct(self):
        axes1=Axes()
        self.add(axes1)

        f1=lambda x: x**2+1

        grafico_f1=axes1.plot(f1,x_range=[0,2],color=PURE_GREEN)

        self.play(Create(grafico_f1))

        self.wait(1)

        x_v=1
        Pgrafico=Dot(axes1.c2p(x_v,f1(x_v)),radius=0.05,color=YELLOW_C)

        self.add(Pgrafico)
        self.wait(1)

        #Etiquetando o valor (x_v,f(x_v))
        Etiqueta_Pgrafico=Tex(f"({x_v:.2f},{f1(x_v):.1f})").scale(0.5)
        #:.2f número de casas decimais
        self.add(Etiqueta_Pgrafico)
        Etiqueta_Pgrafico.next_to(Pgrafico,UP)

        self.wait(2)

        # Voltar na Projeção Horizontal e Vertical
````

#4.8 Adicionando linhas e pontos associados a figuras

Tomaremos um círculo para melhor fixação das ideias.

**1. Associando um ponto ao círculo:**
* Criamos um círculo:

     `C1=Circle()`

* Posicionamos um ponto sobre o círculo no ângulo $\frac{\pi}{4}$:

   `Ponto = Dot().move_to(C1.point_at_angle(PI/4))`

O método `.point_at_angle()` retorna a posição de um ponto na borda do círculo (ou arco) em um ângulo específico.

**Observação:** esse método só funciona para formas circulares, como `Circle` e `Arc`.

**2. Obtendo as coordenadas de um ponto:**

Se estivermos trabalhando com eixos:

`coordenadas = ax.p2c(P2.get_center())`

`P2_x = coordenadas[0]`

`P2_y = coordenadas[1]`

O método `ax.p2c()` converte um ponto da cena em coordenadas $(x,y)$ no sistema de eixos `ax`. Assim conseguimos identificar a posição do ponto em termos do plano cartesiano.

**3. Diferença entre `c2p(x, y)` e `p2c(point)`:**

* `ax.c2p(x, y)`: converte coordenadas cartesianas $(x,y)$ em um ponto no espaço da cena.

* `ax.p2c(point)`: converte um ponto da cena para as coordenadas cartesianas $(x,y)$ correspondentes.

Esses dois métodos são complementares e permitem trabalhar de forma flexível entre os pontos gráficos do Manim e o sistema de coordenadas matemáticas.


````python
# @title
%%manim -qm figurasponto

class figurasponto(Scene):
    def construct(self):
        ax=Axes(
           x_range=[-1,5,1],
           y_range=[-1,5,1],
           x_length= 6,
           y_length=6

        )
        self.add(ax)

        x_0=0
        y_0=1

        centro=ax.c2p(x_0,y_0)

        C1=Circle(radius=1.5*ax.get_x_unit_size(),color=PURE_GREEN).move_to(centro)
        self.play(Create(C1))
        self.wait(2)
        P1=Dot().move_to(centro)

        # Podemos trocar ânglo
        P2=Dot(color=YELLOW_C).move_to(C1.point_at_angle(PI/4))

        #Etiquetar
        et_P1=MathTex(r"(a,b)",color=WHITE).next_to(P1,DOWN*0.8).scale(0.9)
        et_P2=MathTex(r"(x,y)",color=YELLOW_C).next_to(P2,UP*0.8).scale(0.9)

        self.add(P1,P2,et_P1,et_P2)
        self.wait(2)

        # Projeções do ponto P1 no eixo x e y
        L1=DashedLine(ax.c2p(x_0,0),P1)
        L2=DashedLine(ax.c2p(0,y_0),P1)
        self.play(Create(L1),Create(L2))
        self.wait(2)

        #bProjeções do ponto P2

        coordenadas = ax.p2c(P2.get_center())
        P2_x = coordenadas[0]
        P2_y = coordenadas[1]
        print(P2_x,P2_y)

        L3=DashedLine(ax.c2p(P2_x,0),P2).set_color(YELLOW_C)
        L4=DashedLine(ax.c2p(0,P2_y),P2).set_color(YELLOW_C)

        self.play(Create(L3),Create(L4))
        self.wait(2)

        raio=Line(P1,P2).set_color(YELLOW_C)
        self.play(Create(raio))
        self.wait(2)


````

# 4.9 Adicionando Reta Tangente ao Gráfico de uma Função em um Ponto

Iremos adicionar a reta tangente ao gráfico de uma função $f(x)$ usando sua equação:

$$y=f'(x_0)(x-x_0)-f(x_0).$$

Com isso, conseguimos facilmente passear de uma reta tangente a outra apenas mudando o ponto $x_0$.

````python
# @title
%%manim -qm Tan
class Tan(Scene):
    def construct(self):
        ax = Axes(
            x_range=[-3, 3],
            y_range=[-1.5, 1.5],
            x_length=6,
            y_length=3,
            axis_config={"include_numbers": True},
        ).scale(2)

        func = lambda x:np.cos(x)

        #0.00001=1e-5  melhor f'(x)=lim_{h->0}\frac{f(x+h)-f(x-h)}{2h}

        dfunc=lambda x: (func(x + 1e-5) - func(x - 1e-5)) / (2 * 1e-5) #Aproximamos a derivada avaliando a taxa de variação em torno do ponto.

        x0=1
        grafico = ax.plot(func, x_range=(-3,3), color=BLUE)

        ponto_de_tangencia=Dot(ax.c2p(x0, func(x0)), color=ORANGE)
        etiqueta_pt=Tex(f"({x0:.2f},{func(x0):.2f})").next_to(ponto_de_tangencia,UP)
        self.add(etiqueta_pt)

        #Definir reta tangente em (x0,f(x0)): y-f(x0)=f'(x0)(x-x0)
        cof_angular =  dfunc(x0)                    #f'(x_0)
        y_intercept = func(x0)- cof_angular * x0              # f(x)-f'(x)x
        tangente_func = lambda x_val: cof_angular * x_val + y_intercept

        graf_tangente=ax.plot(tangente_func,x_range=[-4,4],color=YELLOW)

        #Linhas ligando os pontos
        L1=DashedLine(ax.c2p(x0,0),ponto_de_tangencia)
        L2=DashedLine(ax.c2p(0,func(x0)),ponto_de_tangencia)



        self.play(Create(ax), Create(grafico))
        self.play(Create(ponto_de_tangencia))
        self.play(Create(graf_tangente))
        self.play(Create(L1),Create(L2))
        self.wait(2)



````

Para representar os ângulos internos de um polígono, usamos a classe `Angle, que cria um arco entre duas retas (ou segmentos).

````python
# @title
%%manim -qm Angulo2

class Angulo2(Scene):
    def construct(self):
        #ax=Axes().add_coordinates()
        #self.add(ax)
        Pol2=RegularPolygon(n=7, color=RED).scale(2)
        #Pol2=RegularPolygon(n=7, color=RED).scale(2).move_to(ax.c2p(0,1))

        self.play(Create(Pol2))
        self.wait(2)


        vertices=Pol2.get_vertices()
        v1=vertices[0]
        v2=vertices[1]
        v3=vertices[2]

        L1=Line(v2,v1).set_color(YELLOW_C)
        L2=Line(v2,v3).set_colo(PURE_CYAN)

        a=Angle(L2,L1, radius=0.5).set_color(PURE_GREEN)

        self.play(Create(L1))
        self.play(Create(L2))
        self.play(Create(a))
        self.wait(2)


        # Crie etiqueta para cada vértice e ânglo
        etiqueta = MathTex(r"V_2", color=RED).scale(0.5)
        etiqueta.next_to(vertices[1], vertices[1] - Pol2.get_center())  # Posicionamento
        self.add(etiqueta)
        self.wait(2)

        a_etiqueta=MathTex(r"\theta_{1}").move_to(a.get_center()).scale(0.6)

        self.add(a_etiqueta)
        self.wait(2)

        # pega o centro


        centro = Pol2.get_center()
        P_alto=Pol2.get_top()

        # marca visual
        ponto = Dot(centro, color=RED)
        ponto_tp=Dot(P_alto, color=BLUE)
        self.add(ponto)
        self.add(ponto_tp)

        texto = Text("Centro").scale(0.5).next_to(ponto, UP)
        texto_tp = Text("Ponto mais alto").scale(0.5).next_to(ponto_tp, UP)

        self.add(texto_tp)

        self.play(Create(Pol2))
        self.play(FadeIn(ponto), Write(texto))

        self.wait()






````
