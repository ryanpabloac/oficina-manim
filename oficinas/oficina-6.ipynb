# 6.2 Utilizando `always_redraw`

No Manim, a função `always_redraw` é utilizada para criar um objeto visual (um Mobject dinâmico). Ela instrui o motor da renderização a reconstruir o objeto do zero em cada frame da animação.

A seguir, vamos aplicar o `always_redraw` em casos práticos para entender seu comportamento na renderização.

## 6.2.1 Ponto móvel sobre a  circunferência

```python
# @title
%%manim -qm circulo

class circulo(Scene):
  def construct(self):
     # Configurar eixos
        axes1 = Axes(
            x_range=[-3, 3,1], # Intervalo do eixo x (mínimo, máximo, passo)
            y_range=[-3, 3, 1], # Intervalo do eixo y (mínimo, máximo, passo)
            x_length=6, # Comprimento do eixo X
            y_length=6, # Comprimento do eixo Y
            axis_config={"color": WHITE},  # Cor dos eixos
            ).scale(1).move_to(LEFT*3).add_coordinates()
        self.add(axes1)
        raio=2*axes1.get_x_unit_size()


        C1=Circle(radius=raio).move_to(axes1.c2p(0,0))
        self.play(GrowFromCenter(C1))
        self.wait(2)
         # Ângulo theta
        theta = ValueTracker(0)
        # Pontos no círculo
        def P(x):
            return Dot(axes1.c2p(raio*np.cos(x),raio*np.sin(x))).set_color(PURE_GREEN)
        def P_x(x):
            return Dot(axes1.c2p(raio*np.cos(x),0)).set_color(YELLOW_C)
        def P_y(x):
            return Dot(axes1.c2p(0,raio*np.sin(x))).set_color(GREEN)

        # Ponto móvel na circunferência

        P2=always_redraw(lambda:P(theta.get_value()))

        # Vamos criar Linhas e raio movél
        # Pode usar P2.get_center(), porém fica ruim

        L_raio=Line(axes1.c2p(0,0),P(PI/4),color=PINK)
        self.add(L_raio)

        self.add(P2)
        self.play(theta.animate.set_value(2 * PI), run_time=5, rate_func=linear)
        self.wait()
        self.play(theta.animate.set_value(PI/2), run_time=4, rate_func=linear)
        self.wait()
```

## 6.2.2 Ponto móvel sobre o gráfico de uma função

```python
# @title
%%manim -ql circulo

class circulo(Scene):
  def construct(self):
     # Configurar eixos
        axes1 = Axes(
            x_range=[-7, 7,1], # Intervalo do eixo x (mínimo, máximo, passo)
            y_range=[-1.5, 1.5], # Intervalo do eixo y (mínimo, máximo, passo)
            x_length=14, # Comprimento do eixo X
            y_length=3, # Comprimento do eixo Y
            axis_config={"color": PURE_GREEN},  # Cor dos eixos
            ).add_coordinates()
        self.add(axes1)


        # Ângulo theta
        x_val = ValueTracker(-6)

        def f1(x):
            return np.cos(3*x)

        # Ponto do gráfico
        def Pg(x):
            return Dot(axes1.c2p(x,f1(x))).set_color(RED)

        #Pontos projeções
        def P_x(x):
            return Dot(axes1.c2p(x,0)).set_color(YELLOW_C)
        def P_y(x):
            return Dot(axes1.c2p(0,f1(x))).set_color(PINK)

        # grpafico de f
        grafico=axes1.plot(f1, x_range=[-6, 6], color=YELLOW_C)
        self.add(grafico)

        ponto_g_movel=always_redraw(lambda:Pg(x_val.get_value()))
        ponto_x_movel=always_redraw(lambda:P_x(x_val.get_value()))
        ponto_y_movel=always_redraw(lambda:P_y(x_val.get_value()))
        self.add(ponto_g_movel,ponto_x_movel,ponto_y_movel)

        #Linhas Ponto gráfico projeões

        line_x=always_redraw(lambda:DashedLine(P_x(x_val.get_value()),Pg(x_val.get_value()))).set_color(YELLOW_C)
        line_y=always_redraw(lambda:DashedLine(P_y(x_val.get_value()),Pg(x_val.get_value()))).set_color(RED)

        self.add(line_x,line_y)
        self.play(x_val.animate.set_value(6), run_time=10, rate_func=linear)
        self.wait()
```

## 6.2.4 Reta Tangente

```python
# @title
%%manim -qm Tangente

class Tangente(Scene):
    def construct(self):
        ax = Axes().add_coordinates()

        def seno(x):
          return np.sin(x)
        G_seno = ax.plot(seno, color=RED)
        alpha = ValueTracker(0)
        #No Manim, o método point_from_proportion(alpha) retorna o ponto que está em uma
        #determinada proporção do caminho percorrido por um objeto.
        point = always_redraw(
            lambda: Dot(
                G_seno.point_from_proportion(alpha.get_value()),
                color=BLUE
            )
        )
        tangent = always_redraw(
            lambda: TangentLine(
                G_seno,
                alpha=alpha.get_value(),
                color=YELLOW,
                length=4
            )
        )
        self.add(ax, G_seno, point, tangent)
        self.play(alpha.animate.set_value(1), rate_func=linear, run_time=10)
```

## 6.2.5 Gráficos de Funções no Plano Polar

```python
# @title
%%manim -qm Polar

class Polar(Scene):
    def construct(self):
        #plano Polar
        axes1 = PolarPlane( ).add_coordinates().scale(0.8)
        self.add(axes1)

        # Definir as funções
        #Função
        def polar_function1(x, k):
            return 1-k*np.sin(np.sin(k*x))


        def polar_function2(x,k):
            return 2*np.sin(1.2*k*x)**2+2*np.cos(k*x)**3

        def polar_function3(x,k):
            return 2.5*np.cosh(np.cos(2*k*x)*np.sin(k*x))

        def polar_function4(x, k):
            return 2.9+np.sin(3*k*x)*np.sin(k*x/32)

        # Rastreador para o parâmetro k
        argumento_tracker = ValueTracker(0.5)  # Inicializa o rastreador com valor 0.5



        # Criação da curva f1
        f1 = always_redraw(lambda:ParametricFunction(lambda t:
            axes1.polar_to_point(polar_function4(t, argumento_tracker.get_value()),t),
            t_range=np.array([0, 12 * PI])
        ).set_color_by_gradient(ORANGE))
        self.add(f1)

        self.play(argumento_tracker.animate.set_value(2), run_time=10, rate_func=smooth)

        self.wait(1)


```

# 6.3 Movimento de Câmera no Manim

No Manim, o movimento de câmera depende de qual versão/estilo de cena você está usando:

* Scene comum (2D) → câmera fixa por padrão.

* MovingCameraScene → câmera pode mover, dar zoom, seguir objetos
* ThreeDScene → câmera pode orbitar, inclinar, girar em 3D

1. Fixar a câmera:
   self.camera.frame.scale(1)
   1=normal

2. Zoom da câmera
self.play(

)

3. move + zoom juntos
self.play(self.camera.frame.animate.move_to(MObjeto),
run_time=3,
rate_func=smooth ou linear ou there_and_back
).scale(valor_do_zoom)

**Observação:** Use

Objeto P1 não seja afetado pelo movimento da câmera
        

```python
# @title
%%manim -qm Zoom_objeto

class Zoom_objeto(MovingCameraScene):
    def construct(self):
        ax = Axes(
            x_range=[0, 3.5*PI],
            y_range=[-4, 4],
            x_length=7
        ).scale(1)
        ax_labels = ax.get_axis_labels()
        ax_group = VGroup(ax, ax_labels)
        self.add(ax_group)

        # texto acompanha câmera
        #texto = Text("Pontos de Máximo")

        #texto.add_updater(lambda m: m.move_to(self.camera.frame.get_center()))

        #self.add(texto)

        # fixar a no seu zoom: Vai mudando a escala para entender:
        # 1=normal
        #zoom>1 diminuir  e zoom<1 aumenta
        self.camera.frame.scale(1.5)


        def f(x):
          return np.sin(x)

        f_1= ax.plot(f, x_range=[0, 3.5*PI, 0.01], color=BLUE)
        self.add(f_1)

        P1=Dot(ax.c2p(PI/2,f(PI/2))).set_color(ORANGE)
        P2=Dot(ax.c2p(3*PI/2,f(3*PI/2))).set_color(RED)
        P3=Dot(ax.c2p(3*PI,f(3*PI))).set_color(YELLOW)
        self.add(P1,P2,P3)
        self.wait(2)

        # Move a camera para o ponto1 dar um zoom
        self.play(self.camera.frame.animate.move_to(P1).scale(0.3),run_time=4, rate_func=smooth)
        # Matém a câmera no mesmo zoom
        self.play( self.camera.frame.animate.move_to(P2),run_time=4, rate_func=smooth)

        # Se diminuir/aumentar a escala ele dar mais zoom em relação a posiçao da câmera anterior
        self.play( self.camera.frame.animate.move_to(P3).scale(1.5),run_time=4, rate_func=smooth)

```

## 6.4.1 Movendo a câmera acompandando o ponto

```python
# @title
%%manim -qh Zoom_objeto1

class Zoom_objeto1(MovingCameraScene):
    def construct(self):
        ax = Axes(
            x_range=[0, 7*PI],
            y_range=[-4, 4],
            x_length=27
        ).scale(1)
        ax_labels = ax.get_axis_labels()
        ax_group = VGroup(ax, ax_labels)
        self.add(ax_group)

        # fixar a no seu zoom: Vai mudando a escala para entender:
        # 1=normal
        #zoom>1 diminuir  e zoom<1 aumenta
        self.camera.frame.scale(1)

        def f(x):
          return np.sin(x)

        f_1= ax.plot(f, x_range=[0, 7*PI, 0.01], color=BLUE)
        self.add(f_1)

        # Mover a câmera para origem

        self.play(self.camera.frame.animate.move_to(ax.c2p(0,0)),run_time=4, rate_func=smooth)
        self.wait(1)

        x_val=ValueTracker(0)
        def Pg(x):
            return Dot(ax.c2p(x,f(x))).set_color(YELLOW)

        def P_x(x):
            return Dot(ax.c2p(x,0)).set_color(GREEN)

        ponto_grafico=always_redraw(lambda:Pg(x_val.get_value()))

        self.add(ponto_grafico)

        # Usar Camera com updater trocar ponto_gráfico

        P_direta=always_redraw(lambda:P_x(x_val.get_value()))
        self.add(P_direta)

        # Trocar o ponto por ponto_grafico.get_center()

        self.camera.frame.add_updater(
    lambda m: m.move_to(P_direta.get_center())
)

        self.play(x_val.animate.set_value(7*PI),run_time=10, rate_func=linear)
        self.wait(2)

```

# 6.5 Trajetória Dinâmicas com  `TracedPath`
No Manim, `TracedPath` é uma classe usada para desenhar automaticamente o rastro (trajetória) de um ponto em movimento. Ela é muito útil para visualizar caminhos percorridos por objetos animados.

A estrutura dessa classe é:

`TracedPath(
    traced_point_func,   
    stroke_color=WHITE,  
    stroke_width=2,      
    dissipating_time=1,
    stroke_opacity=1.0  
)`

onde

| Parâmetro           | Função                                              |
| ------------------- | --------------------------------------------------- |
| `traced_point_func` | Uma função que retorna a posição atual do objeto (ex: `dot.get_center`) |
| `stroke_color`      | Define a cor da trajetória                                   |
| `stroke_width`      | Controla a espessura do traço                                  |
| `dissipating_time`  | Se definido (em segundos), o rastro "evapora" gradualmente atrás do objeto, criando um efeito de cauda de cometa.           |
| `stroke_opacity`    | Define o nível de transparência da linha traçada.                                   |


```python
# @title
%%manim -qm Rastro

class Rastro(Scene):
  def construct(self):
      dot = Dot(color=BLUE).scale(1)

      #dot.get_center → informa ao TracedPath qual posição acompanhar.
      #Enquanto o dot se move, o caminho vai sendo desenhado automaticamente.
      rastro = TracedPath(dot.get_center, stroke_color=RED,stroke_width=4, stroke_opacity=1)

      self.add(dot, rastro)
      self.play(dot.animate.shift(RIGHT * 4+2*UP), run_time=3)
      self.wait()
```

#6.6 Animação com `MoveAlongPath`

o `MoveAlongPath` é o comando usado dentro do `self.play()` para fazer um objeto seguir exatamente o desenho de uma linha ou curva.


```python
# @title
%%manim -qm MovOB
class MovOB(Scene):
    def construct(self):
      # Eixos
        axes =Axes(
            x_range=[-5, 5, 1],
            y_range=[-3, 3, 1],

        )
        self.add(axes)


        #Roda o último

        MOB=Circle(radius=2)
        MOB=Ellipse(width= 2, height= 3)
        MOB=Rectangle(width= 2, height= 3)
        MOB=RegularPolygon(n=8).scale(2)
        MOB = Star(n=5).scale(2)
        MOB = ParametricFunction(
    lambda t: np.array([
        16*np.sin(t)**3,
        13*np.cos(t)-5*np.cos(2*t)-2*np.cos(3*t)-np.cos(4*t),
        0
    ])*0.1,
    t_range=[0, TAU]
       )

        #MOB = Square(side_length=3).set_color(PURE_YELLOW)
        #MOB = axes.plot(lambda x:x**2, x_range=[-5,5])

        dot = Dot(stroke_width=0.4,fill_opacity= 1).move_to(MOB.point_from_proportion(0))

        rastro = TracedPath(dot.get_center, stroke_color=PURE_GREEN,stroke_width=5)

        # Retirar MOB
        self.add(dot, rastro)

        self.play(
            MoveAlongPath(dot, MOB),
            run_time=4,
            rate_func=linear
        )

        self.wait()
```

#6.7 Desenhando o  Gráfico de $y=f(x)$

```python
# @title
%%manim -qm GraficoD

class GraficoD(Scene):
    def construct(self):
        # Eixos
        axes = Axes(
            x_range=[-4, 4, 1],
            y_range=[-3, 3, 1],
            axis_config={"include_numbers": True}
        )
        # definir a função
        def func(x):
            return np.sin(2*x)


        # Ponto que se move no gráfico
        x_tracker = ValueTracker(-4)

        moving_dot = always_redraw(
            lambda: Dot(axes.c2p(x_tracker.get_value(), func(x_tracker.get_value())
                ),
                color=YELLOW
            )
        )

        # Rastro do ponto
        trace = TracedPath(
            moving_dot.get_center,
            stroke_color=RED,
            stroke_width=3
        )

        self.add(axes, moving_dot, trace)

        # Animação do parâmetro e do ponto
        self.play(
            x_tracker.animate.set_value(4),
            run_time=6,
            rate_func=linear
        )

        self.wait()
```

###6.8.1 Ponto brilhante com efeito de aura

Vamos solicitar à IA a elaboração de um script em Manim para a geração de um ponto brilhante com efeito de aura luminosa.

```python
# @title
%%manim -qm pontoB

class pontoB(Scene):
  def construct(self):
    dot = Dot(radius=0.08, color=YELLOW)

    grupo_c = VGroup(*[
    Circle(radius=0.15 + i*0.05)
    .move_to(dot)
    .set_fill(YELLOW, opacity=0.15/(i+1))
    .set_stroke(width=0)
    for i in range(6)
])

    ponto_brilhante = VGroup(grupo_c, dot)

    #self.add(ponto_brilhante)
    MOB=Circle(radius=2)
    MOB=Ellipse(width= 2, height= 3)
    MOB=Rectangle(width= 2, height= 3)
    MOB=RegularPolygon(n=8).scale(2)
    MOB = Star(n=5).scale(2)

    MOB = FunctionGraph(lambda x: np.sin(x), x_range=[-5,5])


    ponto_brilhante.move_to(MOB.point_from_proportion(0))

    rastro = TracedPath(
    ponto_brilhante.get_center,
    stroke_color=PURE_GREEN,
    stroke_width=5
)

    self.add(ponto_brilhante, rastro)

    self.play(
    MoveAlongPath(ponto_brilhante, MOB),
    run_time=4,
    rate_func=linear
)



```
