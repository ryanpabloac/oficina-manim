```python
%%manim -qm -v warning Teste2

class Teste2(Scene):
    def construct(self):

        AX1 = Axes(
            x_range=[-PI, PI, PI/4],
            y_range=[-1.5, 1.5, 0.5],
            x_length=5,
            y_length=4,
        ).scale(0.6).to_edge(LEFT)

        AP1 = PolarPlane().to_edge(RIGHT)

        self.add(AX1, AP1)

        seno = AX1.plot(
            lambda x: np.sin(x),
            color=GREEN
        )
        label_seno =AX1.get_graph_label(seno,label=MathTex(r"y = \sin(x)"),x_val=1,direction=UP)
        f1_group = VGroup(seno, label_seno)

        self.play(Create(f1_group), run_time=1)

        pg = ParametricFunction(
            lambda t: AP1.polar_to_point(3*np.sin(2*np.cos(3*t)),t),
            t_range=[0,8*PI],
            color=RED,
        )

        self.play(Create(pg), run_time=3)
        self.wait(2)
```
